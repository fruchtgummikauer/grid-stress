# Grid-load models: our day-ahead grid-load forecast vs. SMARD

A copy of [`regression-models-claude.ipynb`](regression-models-claude.ipynb) (spec 06, after the team's
refactor) with the **target switched from `residual_load` to `grid_load`**. No spec behind it; `-magc`
marks it as a team member's experiment, not adopted.

We forecast `grid_load` for every hour of a delivery day `DAY`, **issued at 18:00 on `DAY−1`**: the
same point in time as SMARD's public day-ahead forecast. Every model is scored against SMARD's
`fc_grid_load` and against a seasonal-naive floor, on **identical hours**.

**Why this notebook exists.** The residual-load models beat SMARD in the tails mostly by correcting
SMARD's *load* forecast, which they take as an input. Here `fc_grid_load` is the **benchmark only and
never an input** (asserted in [§4.6](#sec-4-6) and [§9.1](#sec-9)). If our own load forecast holds up,
`own load − fc_gen_wind_solar` can later replace SMARD's load input in the residual-load model.

**Still one SMARD input, switchable:** SMARD's wind + solar forecast `fc_gen_wind_solar`
(`smard_forecast_renewables`, on by default), because rooftop solar lowers the measured grid load on
sunny days. Switch it off for a forecast that uses no SMARD forecast at all.

## How to use this notebook

- Change a value in the configuration cells ([§1.3](#sec-1-3)), then re-run top to bottom. No later cell
  hardcodes a value the configuration holds. `TARGET`, `BENCHMARK` and `BENCHMARK_ERROR` name the
  forecast series and SMARD's forecast of it.
- The interpretation is written **once**, in the closing section, for the default configuration.
  Tables and plots in between carry titles and units, no commentary.
- `SEARCH` ([§1.3](#sec-1-3)) sets how the boosters and the linear model are tuned on the validation walk-forward ([§3.3](#sec-3-3)): every grid
  configuration (`GridSearchCV`) or `n_iter` draws (`RandomizedSearchCV`). Each of them is built
  as an sklearn pipeline by `build_pipeline` ([§4.5](#sec-4-5)).

## What this notebook produces

- day-ahead grid-load forecasts from seasonal naive `DAY−7` and every registry model switched on in `MODELS`
  ([§1.3](#sec-1-3)), each under a **static** and a **rolling** split method on the same test year
- an empirical 95 % prediction interval per model, with its measured coverage
- one scoreboard (accuracy, extremes, intervals) including SMARD and seasonal naive
- a residual-load forecast built from every load forecast (`own load − fc_gen_wind_solar`), scored
  against SMARD's `fc_residual_load` ([§10](#sec-10))
- two optional exports in `data/models/` (`model_load_*.csv`), written only when `EXPORT_ENABLED` is on (default off)

## Conventions

- **Sign:** `error = forecast − actual`, as in spec 04. **Positive = over-forecast.**
- **Units:** hourly readings, forecasts and errors in `MWh`; capacity in `MW`; skill and coverage in
  `%`.
- **Durations, never row counts.** No literal calendar year or date appears in code.
- `time_series` holds exactly `SERIES + DERIVED`. Everything this notebook builds lives in separate
  frames.

## Not in this notebook
- risk flags on our forecast (parked [04.3](../../.claude/specs/04.3-risk-label-link.md))

### 1.2 Resolution and durations

Every window, lag, refit interval, Fourier period and issue time in this notebook is a
**duration** (`pd.Timedelta`), never a number of rows:

- `RESOLUTION` is measured from the data: the most common step between two rows
- `n_obs(duration)` turns a duration into a number of rows where one is needed (rows per day)
- a day counts as complete with at least `DAY_COMPLETENESS` (23 of 24) of its rows, so the
  23-hour spring-DST day is accepted

> if SMARD's data switched from hourly to 15-minute values, no setting would have to change

- models could then be compared **with each other** at 15 minutes
- and with SMARD's forecast only after **averaging to hours**: `smard_forecast_errors_hourly.csv`
  holds hourly errors

````python
# Time step between two rows, measured from the data (the most common gap)
RESOLUTION = time_series.index.to_series().diff().mode().iloc[0]

# A day is complete with at least 23 of 24 hours: accepts the spring-DST day
DAY_COMPLETENESS = 23 / 24


def n_obs(duration):
    """Number of rows `duration` spans at RESOLUTION; stops if it is not a whole number."""
    count = duration / RESOLUTION
    if count != int(count):
        raise ValueError(
            f"{duration} is not a whole multiple of the resolution {RESOLUTION}"
        )
    return int(count)


def describe(value):
    """Short readable text for a setting, e.g. '24h', '7 days' or '24 months'."""
    if isinstance(value, pd.Timedelta):
        whole_days = value == pd.Timedelta(days=value.days)
        if value >= pd.Timedelta(days=2) and whole_days:
            return f"{value.days} days"
        hours = value / pd.Timedelta(hours=1)
        return f"{hours:g}h"

    if isinstance(value, pd.DateOffset):
        parts = []
        for unit, amount in value.kwds.items():
            if amount == 1:
                unit = unit.rstrip("s")  # "1 month", not "1 months"
            parts.append(f"{amount} {unit}")
        return ", ".join(parts)

    return str(value)


EXPECTED_OBS_PER_DAY = n_obs(pd.Timedelta(days=1))
MIN_OBS_PER_DAY = int(np.ceil(DAY_COMPLETENESS * EXPECTED_OBS_PER_DAY))

print(
    f"resolution {describe(RESOLUTION)}: {EXPECTED_OBS_PER_DAY} observations per full day"
)
print(
    f"a day is complete with >= {MIN_OBS_PER_DAY} of {EXPECTED_OBS_PER_DAY} observations "
    "(accepts the spring-DST day)"
)
````

````python
# Registry helpers (the §1.5 self-check also uses is_distribution and grid_size)
def is_distribution(value):
    """True for a scipy.stats distribution, False for a list of values."""
    return hasattr(value, "rvs")


def grid_size(grid):
    """Number of configurations in `grid`; None if a value is a distribution."""
    if any(is_distribution(values) for values in grid.values()):
        return None
    return int(np.prod([len(values) for values in grid.values()]))


def configurations_tried(grid):
    """Number of configurations the search evaluates for `grid` under SEARCH."""
    size = grid_size(grid)
    if SEARCH["method"] == "grid" or not grid:
        return size
    if size is None:
        return SEARCH["n_iter"]
    return min(SEARCH["n_iter"], size)


def show_grid(grid):
    """Readable form of `grid`: lists as they are, distributions as name(args)."""
    if not grid:
        return "NA"
    shown = {}
    for name, values in grid.items():
        if is_distribution(values):
            shown[name] = f"{values.dist.name}{values.args}"
        else:
            shown[name] = values
    return shown


# How the search is set up
search_method = SEARCH["method"]
if search_method == "random":
    search_method += f" ({SEARCH['n_iter']} draws, seed {SEED})"
print("SEARCH")
print(f"  method : {search_method}")
print(f"  n_jobs : {SEARCH['n_jobs']}")
print(f"  verbose: {SEARCH['verbose']}")

# One row per model
rows = []
for model_key, model in MODELS.items():
    rows.append(
        {
            "model": model_key,
            "enabled": model["enabled"],
            "family": model["family"],
            "grid configurations": grid_size(model["grid"]) or "unbounded",
            "configurations tried": configurations_tried(model["grid"]),
            "grid": show_grid(model["grid"]),
            "label": model["label"],
            "color": model["color"],
        }
    )
registry = pd.DataFrame(rows).set_index("model")
display(registry)

# The fixed parameters of each model (SARIMAX's Fourier periods shown as e.g. '24h')
print("fixed parameters")
for model_key, model in MODELS.items():
    params = dict(model["params"])
    if "fourier" in params:
        params["fourier"] = {
            describe(period): order for period, order in params["fourier"].items()
        }
    print(f"  {model_key:<16}: {params}")
````

## 2 Forecast setting

Every model and every training row obeys one rule for **what is known when**. Defaults from
`DATA_INFO` in brackets:

| Item | Rule |
|---|---|
| issue time | the forecast for delivery day `DAY` is issued on `DAY−1` (18:00 local) |
| availability cutoff | issue time − actuals lag (2 h, so 16:00 on `DAY−1`) |
| actuals | a row stamped `t` (interval start) is usable if `t + resolution ≤ availability cutoff`: up to the row stamped 15:00 |
| SMARD forecasts for `DAY` | available: both components are published by 18:00 on `DAY−1` |
| SMARD forecasts for the rest of `DAY−1` | available: published on `DAY−2` |
| target hours | every local hour of `DAY`: 24, or 23 on the spring DST day (the autumn fold is collapsed by SMARD) |
| training rows | delivery days whose target hours are all observed by the fit's cutoff, each row built as it would have been at its own issue time |

Issue time and availability cutoff are **per-day timestamps** (`FORECAST_SETTING`'s `issue_time`
and `cutoff`), never constants. The truncation leakage test that proves the rule holds needs the feature builder and
the fitted models, so it runs in [§5.4](#sec-5-4).


`delivery day`: 
- ``DAY`` as a forecast target = the forecast day for which the power will be *delivered*
- model trains on **earlier** days up to ``DAY-2``
- we can't use ``DAY-1`` because its actuals (the training target) from 16:00 on are not observed by the cutoff

### 2.1 The forecast setting

- `FORECAST_SETTING` holds, per ``delivery day``:
  - `issue_time`
  - `cutoff` (the availability cutoff)
  - `target_hours`

- `ROW_FORECAST_SETTING` the same times per row
- `available(stamps, cutoff)` applies the actuals rule above.

````python
# Delivery day of every row: its local midnight
DAY_OF_ROW = time_series.index.normalize()
TARGET_HOURS = time_series.index.groupby(DAY_OF_ROW)  # delivery day -> its local hours
DAYS = pd.DatetimeIndex(sorted(TARGET_HOURS))

# Forecast setting per delivery day
FORECAST_SETTING = pd.DataFrame(index=DAYS)
FORECAST_SETTING["issue_time"] = (
    DAYS - DATA_INFO["issue_days_before"] + DATA_INFO["issue_clock"]
)
FORECAST_SETTING["cutoff"] = FORECAST_SETTING["issue_time"] - DATA_INFO["actuals_lag"]
FORECAST_SETTING["target_hours"] = [TARGET_HOURS[day] for day in DAYS]

# The same times per row
ROW_FORECAST_SETTING = pd.DataFrame({"day": DAY_OF_ROW}, index=time_series.index)
ROW_FORECAST_SETTING = ROW_FORECAST_SETTING.join(
    FORECAST_SETTING[["issue_time", "cutoff"]], on="day"
)


def available(stamps, cutoff):
    """True where a row stamped `stamps` (interval start) has ended by `cutoff`."""
    return stamps + RESOLUTION <= cutoff


hours_per_day = FORECAST_SETTING["target_hours"].map(len)
dst_days = FORECAST_SETTING.index[hours_per_day < EXPECTED_OBS_PER_DAY]

print(f"Delivery days   : {len(DAYS):,} ({DAYS[0]:%Y-%m-%d} .. {DAYS[-1]:%Y-%m-%d})")
print(
    f"Spring DST days : {len(dst_days)}, "
    f"each with {hours_per_day[dst_days].min()} target hours"
)
print("\nExample: the latest spring DST day and the record's last day")
display(FORECAST_SETTING.loc[[dst_days[-1], DAYS[-1]], ["issue_time", "cutoff"]])
````

### <a id="sec-2-2"></a>2.2 Feature availability rules

| Input | Usable at the issue time | Why |
|---|---|---|
| actuals: `grid_load`, wind + solar generation (`renewables`), and SMARD's errors (they need the actual) | rows with `t + resolution ≤ availability cutoff` | actuals appear late on smard.de  |
| SMARD component forecasts for the target hour | yes | published by 18:00 on `DAY−1` |
| SMARD component forecasts for `DAY−1` from the cutoff on | yes | published on `DAY−2`; SARIMAX's exogenous inputs |
| calendar and cyclical: hour, weekday, month, weekend, holiday; sine / cosine of time of day, weekday and day of year; season | always | known in advance |
| installed capacity | the latest value published by the issue time; the capacity factor (`renewables_history`) uses the value at its `DAY−2` hour, published by then | publication rule, [§2.3](#sec-2-3) |

The cell below checks the same-hour lags in `FEATURE_WINDOWS` against the rule, for every row of the
record, and prints each lag's smallest margin to the cutoff. A lag that reaches past the cutoff (e.g.
`DAY−1`, whose afternoon and evening come after 16:00) stops the notebook instead of leaking
silently. The `renewables_history` columns use the first same-hour lag (`DAY−2`), so this check covers
them; the `rolling_stats` windows end at the cutoff by definition. The leakage test
([§5.4](#sec-5-4)) checks every built feature.

- **The 2-day lag is the tightest: its source rows end at least 16h before the cutoff.**
  - The 7-day lag keeps at least 136h.

````python
# Stop if a same-hour lag reaches past the availability cutoff
cutoffs = ROW_FORECAST_SETTING["cutoff"].to_numpy()

for lag in FEATURE_WINDOWS["same_hour_lags"]:
    source = time_series.index - lag
    usable = available(source, cutoffs)
    if not usable.all():
        raise ValueError(
            f"same-hour lag {describe(lag)} reaches past the availability cutoff for "
            f"{(~usable).sum():,} rows: it would leak. "
            "Use a lag the forecast setting allows."
        )
    margin = (cutoffs - (source + RESOLUTION)).min()
    print(f"lag {describe(lag)}: smallest margin to the cutoff {describe(margin)}")
````

### <a id="sec-2-3"></a>2.3 Capacity publication rule

- `cap_*` columns hold one value per calendar year
  - we treat year `Y`'s value as **published on 1 January of `Y`, 00:00** (`DATA_INFO`)
  - A delivery day uses the latest value published at or before its issue time:

- Most delivery days use the **current** calendar year's value.
- **1 January uses the previous year's value.** It is issued at 18:00 on 31 December, before the new
  year's value is out.
- Only the record's first day is issued before the first publication and has **no** capacity value.
  It is unforecastable anyway ([§2.4](#sec-2-4)), and the default 24-month windows never reach it.

Growth of the installed power plant capacity within a year is invisible with the SMARD data source!

````python
CAP_COLUMNS = ["cap_wind_off", "cap_wind_on", "cap_solar"]


def capacity_table(frame):
    """Per calendar year: the cap_* values, their total (MW) and publication time."""
    per_year = frame[CAP_COLUMNS].groupby(frame.index.year)
    if (per_year.nunique() > 1).any().any():
        raise ValueError(
            "a cap_* column changes within a calendar year: "
            "the yearly-step assumption fails"
        )
    table = per_year.first().rename_axis("capacity_year")
    table["cap_total"] = table[CAP_COLUMNS].sum(axis=1)
    delay = DATA_INFO["capacity_published_after"]
    table["published_at"] = [
        pd.Timestamp(year=year, month=1, day=1) + delay for year in table.index
    ]
    return table


def published_capacity(issue_times, table):
    """Capacity year and total (MW) of the latest value published by each issue time.

    Empty (NaN) before the first publication.
    """
    by_publication = table.reset_index().set_index("published_at")
    latest = by_publication[["capacity_year", "cap_total"]].reindex(
        pd.DatetimeIndex(issue_times), method="ffill"
    )
    return latest.reset_index(drop=True).astype({"capacity_year": "Int64"})


CAPACITY = capacity_table(time_series)
DAY_CAPACITY = published_capacity(FORECAST_SETTING["issue_time"], CAPACITY)
DAY_CAPACITY = DAY_CAPACITY.set_axis(FORECAST_SETTING.index)

print("Installed capacity per year (MW) and its publication time")
display(CAPACITY.style.format("{:,.0f}", subset=CAP_COLUMNS + ["cap_total"]))

# The record's latest 1 January and its two neighbours
new_year = DAYS[(DAYS.month == 1) & (DAYS.day == 1)][-1]
around = [new_year - pd.Timedelta(days=1), new_year, new_year + pd.Timedelta(days=1)]
example = FORECAST_SETTING.loc[around, ["issue_time"]].join(DAY_CAPACITY)

print(f"\n1 January ({new_year:%Y-%m-%d}) still uses the previous year's value")
display(example.style.format({"cap_total": "{:,.0f}"}))
````

### <a id="sec-2-4"></a>2.4 Unforecastable days

`DAY` cannot be forecast by **any** model as soon as:
 - a SMARD component forecast (`fc_grid_load`, `fc_gen_wind_solar`) is missing for an
   - **hour** of `DAY`
   - or for an **hour** of `DAY−1` from the cutoff on.
- it is left out of training and scoring and counted here
- the record's first day is excluded for the same reason: its `DAY−1` lies outside the record.

No model here reads `fc_grid_load`, but it stays in this rule: it is the benchmark every row is
scored against, and keeping the rule keeps the forecastable days identical to
`regression-models-claude.ipynb`.

````python
FORECAST_INPUTS = ["fc_grid_load", "fc_gen_wind_solar"]

# Row ranges per delivery day: DAY−1 from the cutoff on, and DAY itself
missing_input = time_series[FORECAST_INPUTS].isna().any(axis=1).to_numpy()
first_unavailable = time_series.index.searchsorted(FORECAST_SETTING["cutoff"])
day_start = time_series.index.searchsorted(FORECAST_SETTING.index)
day_end = time_series.index.searchsorted(FORECAST_SETTING.index + pd.Timedelta(days=1))

EXCLUSION = pd.DataFrame(
    {
        "forecast missing on DAY": [
            missing_input[start:end].any() for start, end in zip(day_start, day_end)
        ],
        "forecast missing on DAY−1 after the cutoff": [
            missing_input[start:end].any()
            for start, end in zip(first_unavailable, day_start)
        ],
        "DAY−1 outside the record": FORECAST_SETTING["cutoff"] < time_series.index[0],
    },
    index=FORECAST_SETTING.index,
)
FORECASTABLE = ~EXCLUSION.any(axis=1)

print(f"Unforecastable delivery days: {(~FORECASTABLE).sum()} of {len(FORECASTABLE):,}")
excluded = EXCLUSION[~FORECASTABLE]
reasons = excluded.apply(lambda row: ", ".join(row.index[row]), axis=1)
display(reasons.rename("reason").rename_axis("DAY").to_frame())
````

## 3 Windows and split methods

- **Test window:**
  - the 365 delivery days ending with the last **complete** day: the last day whose hours all carry the actual and every SMARD forecast
- **Validation window:** the 365 delivery days before it. Both are derived from the data.
- **Training rows of a fit:**
  - the delivery days inside the 24-month window ending at the fit's cutoff whose target hours are all observed by that cutoff (up to `DAY−2` for a fit issued on `DAY−1`)
  - Unforecastable days and days with a missing actual are left out.
- **Tuning:**
  - rolling walk-forward over the validation year
  - as sklearn's `GridSearchCV` or `RandomizedSearchCV` (`SEARCH`) with one fold per validation refit
  - The configuration with the lowest validation MAE is selected and **frozen**: both split methods reuse it unchanged, and refits never re-run the search.
- **Static:**
  - one fit issued for the first test day
  - It forecasts the whole test year with daily updated inputs and frozen parameters.
- **Rolling:**
  - the same first fit, then a refit every `WINDOWS["refit_every"]` on the sliding 24-month window
  - Until the first refit, both split methods forecast identically.
- **Test-year fits** (static and rolling) train on windows that **include the validation year**:
  tuning only chose the configuration.

Both split methods forecast **identical test hours** with equal window lengths, so the only difference between them is the refitting.

### 3.1 Test and validation windows

````python
JOINT_COLUMNS = [
    TARGET,
    BENCHMARK,
    "fc_gen_wind_solar",
]

# Per delivery day: its rows, and whether the actual / every joint column is observed
DAY_INFO = (
    time_series.index.to_series()
    .groupby(DAY_OF_ROW)
    .agg(rows="size", first_row="min", last_row="max")
)
actual_observed = time_series[TARGET].notna()
joint_observed = time_series[JOINT_COLUMNS].notna().all(axis=1)
DAY_INFO["actual_complete"] = actual_observed.groupby(DAY_OF_ROW).all()
DAY_INFO["jointly_observed"] = joint_observed.groupby(DAY_OF_ROW).all()
assert DAY_INFO.index.equals(DAYS)

# Full clock span: 00:00 to 23:00 and complete (the spring-DST day passes)
full_span = (
    (DAY_INFO["first_row"] == DAY_INFO.index)
    & (DAY_INFO["last_row"] == DAY_INFO.index + pd.Timedelta(days=1) - RESOLUTION)
    & (DAY_INFO["rows"] >= MIN_OBS_PER_DAY)
)
TRAINABLE = FORECASTABLE & full_span & DAY_INFO["actual_complete"]
COMPLETE = full_span & DAY_INFO["jointly_observed"]

TEST_END = COMPLETE.index[COMPLETE][-1]
TEST_DAYS = DAYS[(DAYS > TEST_END - WINDOWS["test"]) & (DAYS <= TEST_END)]
VAL_DAYS = DAYS[(DAYS >= TEST_DAYS[0] - WINDOWS["validation"]) & (DAYS < TEST_DAYS[0])]


def hours_of(days):
    """The record's local hours of the delivery days in `days`."""
    return time_series.index[DAY_OF_ROW.isin(days)]


# One row per window
window_rows = []
for window_name, window_days in [("validation", VAL_DAYS), ("test", TEST_DAYS)]:
    forecastable = FORECASTABLE.loc[window_days]
    no_actual = forecastable & ~TRAINABLE.loc[window_days]
    window_rows.append(
        {
            "window": window_name,
            "first day": f"{window_days[0]:%Y-%m-%d}",
            "last day": f"{window_days[-1]:%Y-%m-%d}",
            "days": len(window_days),
            "hours": len(hours_of(window_days)),
            "unforecastable days": int((~forecastable).sum()),
            "days without a complete actual": int(no_actual.sum()),
        }
    )
print("Test and validation windows (delivery days, local time)")
display(pd.DataFrame(window_rows).set_index("window"))
````

### <a id="sec-3-2"></a>3.2 Fit schedule and training windows

One schedule per run. The two validation runs exist for tuning and for calibrating the prediction
bands ([§6](#sec-6)): the rolling band from the tuning walk-forward, the static band from one frozen fit over
the validation year.

````python
def training_days(fit_day):
    """Keeps & returns only the training days that:
    - fall inside the train window before that fit day's cutoff (16:00 on DAY−1)
    - are fully observed by the cutoff
    - are trainable (full 24 hours, complete actual data)
    (§3's training-row rule).
    """
    cutoff = FORECAST_SETTING.at[fit_day, "cutoff"]
    in_window = DAYS >= cutoff - WINDOWS["train"]
    observed_by_cutoff = available(DAY_INFO["last_row"], cutoff).to_numpy()
    return DAYS[in_window & observed_by_cutoff & TRAINABLE.to_numpy()]


RUNS = {
    "validation_rolling": {         # tuning; calibrates the rolling band
        "split_method": "rolling",
        "days": VAL_DAYS,
        "enabled": True,
    },
    "validation_static": {          # calibrates the static band
        "split_method": "static",
        "days": VAL_DAYS,
        "enabled": TRAIN_TEST_SPLIT_METHOD["static"],
    },
    "test_static": {                # test year, one frozen fit
        "split_method": "static",
        "days": TEST_DAYS,
        "enabled": TRAIN_TEST_SPLIT_METHOD["static"],
    },
    "test_rolling": {               # test year, refit every WINDOWS["refit_every"]
        "split_method": "rolling",
        "days": TEST_DAYS,
        "enabled": TRAIN_TEST_SPLIT_METHOD["rolling"],
    },
}

# Fit issue days: the first day, then one every refit_every (rolling only)
for run_name, run_info in RUNS.items():
    run_days = run_info["days"]
    if run_info["split_method"] == "rolling":
        run_info["fit_days"] = pd.date_range(
            run_days[0], run_days[-1], freq=WINDOWS["refit_every"]
        )
    else:
        run_info["fit_days"] = run_days[:1]

    window_start = (
        FORECAST_SETTING.at[run_info["fit_days"][0], "cutoff"] - WINDOWS["train"]
    )
    if window_start < time_series.index[0]:
        raise ValueError(
            f"{run_name}: the first fit's {describe(WINDOWS['train'])} training window "
            f"starts {window_start:%Y-%m-%d}, before the record "
            f"({time_series.index[0]:%Y-%m-%d}). "
            "Shorten WINDOWS['train'] or re-fetch more history."
        )


def span(days):
    """First and last day of `days`, and their count."""
    return f"{days[0]:%Y-%m-%d} .. {days[-1]:%Y-%m-%d} ({len(days)} days)"


# One row per run
schedule_rows = []
for run_name, run_info in RUNS.items():
    schedule_rows.append(
        {
            "run": run_name,
            "split method": run_info["split_method"],
            "enabled": run_info["enabled"],
            "fits": len(run_info["fit_days"]),
            "forecast days": span(run_info["days"]),
            "first fit: training days": span(training_days(run_info["fit_days"][0])),
            "last fit: training days": span(training_days(run_info["fit_days"][-1])),
        }
    )
print("Fit schedule per run")
display(pd.DataFrame(schedule_rows).set_index("run"))
````

### <a id="sec-3-3"></a>3.3 The split engine

Each model family supplies one function, registered in `FAMILY_FIT` in the Models section:
`fit(model_key, config, train_days, forecast_days)` returns `predict(day)`, which forecasts every target
hour of one delivery day. The engine calls it per scheduled fit and applies the failure rules:

- A fit that **raises** leaves its days empty until the next successful fit. The previous fit does
  not keep forecasting, and no other model fills in.
- A day whose forecast **raises** stays empty.
- A **convergence warning** is not a failure: the forecast is kept and the warning is counted.

`fit_schedule(run)` assigns each forecastable day of a run to the latest fit at or before it.
`run_split` forecasts those days, and `walk_forward_folds` turns the same schedule into sklearn
folds: training rows and forecast hours with an actual, as positions in `FEATURE_TABLE`.

````python
FAMILY_FIT = {}  # family -> fit function, filled in the Models section


def fit_schedule(run):
    """Per fit of `run` return the pairs: (fit_day, all days the fit day forecasts).

    - Each day is forecast by the latest fit at or before it.
    - Static run: one fit day = first day of the window
    - Rolling run: one fit day every WINDOWS["refit_every"] interval (e.g. 30 days)
    - Only forecasteable days are included
    """
    days, fit_days = RUNS[run]["days"], RUNS[run]["fit_days"]
    forecasting_fit = fit_days.searchsorted(days, side="right") - 1
    forecastable = FORECASTABLE.loc[days].to_numpy()
    return [
        (fit_day, days[(forecasting_fit == i) & forecastable])
        for i, fit_day in enumerate(fit_days)
    ]


def run_split(model_key, config, run):
    """Forecast the days of `run` with model `model_key` under `config`.

    Returns:
        - Forecast Series: hourly forecast over the run's hours (NaN where no forecast exists)
        - Logging Dataframe with one row per fit (Status, Timing, Warnings).
    """
    pieces, log = [], []

    # Get pairs for fit schedule (fit_day, forecast_days)
    for fit_day, forecast_days in fit_schedule(run):
        train_days = training_days(fit_day)  # keeps only days inside the training window (protects fit from seeing the future)
        entry = {"fit_day": fit_day, "status": "ok", "failed_days": 0}  # log entry start

        with warnings.catch_warnings(record=True) as caught:
            warnings.simplefilter("always")

            # Fit: a failure leaves this fit's days empty
            start = time.perf_counter()
            try:
                fit = FAMILY_FIT[MODELS[model_key]["family"]]
                predict = fit(model_key, config, train_days, forecast_days)
            except Exception as error:
                predict = None
                entry["status"] = f"failed: {type(error).__name__}: {error}"
            entry["fit_seconds"] = time.perf_counter() - start

            # Forecast day by day: a failed day stays empty
            start = time.perf_counter()
            days_to_forecast = forecast_days if predict is not None else []
            for day in days_to_forecast:
                try:
                    forecast = predict(day)
                except Exception:
                    entry["failed_days"] += 1
                    continue
                target_hours = FORECAST_SETTING.at[day, "target_hours"]
                hourly = forecast.index.equals(target_hours)
                message = f"{model_key}: {day:%Y-%m-%d} is not forecast hour by hour"
                assert hourly, message
                pieces.append(forecast)  # one piece = one valid forecast
            entry["forecast_seconds"] = time.perf_counter() - start

        convergence = sum(issubclass(w.category, ConvergenceWarning) for w in caught) # Only for SARIMAX
        entry["convergence_warnings"] = convergence
        entry["other_warnings"] = len(caught) - convergence
        log.append(entry)

    forecast = pd.concat(pieces) if pieces else pd.Series(dtype=float)
    forecast = forecast.reindex(hours_of(RUNS[run]["days"])).astype(float)
    return forecast, pd.DataFrame(log)

def walk_forward_folds(run):
    """ Builds the 'cv=' argument for GridSearchCV/RandomizedSearchCV
    - only for the CV search to build every candidate configuration
    - does not keep the forecasts
    - forecast rows must have "actual" values for TARGET

    Returns: (training positions(rows), test=forecast positions (rows)).
    """
    has_actual = time_series[TARGET].notna().to_numpy()
    folds = []
    for fit_day, forecast_days in fit_schedule(run):
        train = FEATURE_TABLE.index.get_indexer(hours_of(training_days(fit_day)))
        test = FEATURE_TABLE.index.get_indexer(hours_of(forecast_days))
        test = test[has_actual[test]]
        if len(test):  # a fit without a single scorable hour is no fold
            folds.append((train, test))
    return folds
````

**Tuning.** `tune` selects the configuration with the lowest validation MAE:

- **Boosters and the linear model:** `search` runs `SEARCH` over the folds (`GridSearchCV` or `RandomizedSearchCV`, `refit=False`).
  - A configuration's MAE is **hour-weighted** over the folds, i.e. pooled over every validation hour. sklearn's plain fold mean would weigh the short last fold like a full one.
  - A configuration with a failed fold gets no MAE. If every fit fails, the search stops with sklearn's error.
  - The winner runs once more through `run_split`: the search returns no forecasts, and the rolling band ([§6](#sec-6)) is calibrated on them.
- **SARIMAX** has no search: `tune_sarimax` runs its fixed parameters once through the validation walk-forward.

`evaluate_model` then freezes that configuration and runs every enabled split method.

````python
def search(model_key):
    """SEARCH over the grid of `model_key` on the validation walk-forward.
    - tries every candidate configuration on rolling validation folds

    Returns:
    - every evaluated configuration (full parameter dicts)
    - table with one row per config with it's validation MAE (failed folds have a NaN entry)
    """
    model = MODELS[model_key]
    folds = walk_forward_folds("validation_rolling")
    prefix = PARAM_PREFIX[model["architecture"]]
    space = {prefix + name: values for name, values in model["grid"].items()}
    options = {
        "scoring": "neg_mean_absolute_error",  # Sklearn convention "higher is better" therefore flipping the MAE
        "cv": folds,
        "refit": False,  # We use the search only to look for the best configuration -> we fit later in "run_split()"
        "error_score": np.nan,  # On error search keeps going, but an error fold has no score
        "n_jobs": SEARCH["n_jobs"],
        "verbose": SEARCH["verbose"],
    }
    estimator = build_pipeline(model_key, model["params"])
    if SEARCH["method"] == "grid":
        searcher = GridSearchCV(estimator, space, **options) # ** unpacks the options dict directly here (DRY) | Grid Search: every combination!
    else:
        searcher = RandomizedSearchCV(
            estimator, space, n_iter=SEARCH["n_iter"], random_state=SEED, **options # ** DRY unpack | Randomized Search: max. n_iter draws!
        )
    searcher.fit(FEATURE_TABLE, time_series[TARGET]) # .fit(X, y)
    results = searcher.cv_results_

    # Hour-weighted MAE: NaN as soon as one fold failed
    sizes = np.array([len(test) for _, test in folds]) # scored hours per fold
    fold_scores = [results[f"split{i}_test_score"] for i in range(len(folds))]
    mae_per_fold = -np.column_stack(fold_scores)
    mae = mae_per_fold @ sizes / sizes.sum()  # gives the MAE over all validation hours
    failed = np.isnan(mae_per_fold).sum(axis=1)
    seconds = (results["mean_fit_time"] + results["mean_score_time"]) * len(folds)

    configs, rows = [], []
    for i, point in enumerate(results["params"]):
        tuned = {name.removeprefix(prefix): value for name, value in point.items()} # remove pipeline prefix, plain model names again
        configs.append({**model["params"], **tuned})
        rows.append(
            {
                **{name: tuned[name] for name in model["grid"]},
                "MAE": mae[i],
                "hour_count": int(sizes.sum()) if not failed[i] else 0,
                "failed fits": int(failed[i]),
                "seconds": seconds[i],
            }
        )
    return configs, pd.DataFrame(rows)


def tune_sarimax(model_key):
    """SARIMAX has no search: its fixed parameters on the validation walk-forward."""
    config = dict(MODELS[model_key]["params"])
    forecast, log = run_split(model_key, config, "validation_rolling")
    actual = time_series[TARGET]
    hours = forecast.dropna().index.intersection(actual.dropna().index)
    row = {
        "MAE": (forecast[hours] - actual[hours]).abs().mean(),
        "hour_count": len(hours),
        "failed fits": int((log["status"] != "ok").sum()),
        "seconds": log[["fit_seconds", "forecast_seconds"]].to_numpy().sum(),
    }
    usable = pd.notna(row["MAE"])
    return {
        "table": pd.DataFrame([row]),
        "selected": config if usable else None,
        "forecast": forecast if usable else None,
        "log": log if usable else None,
        "seconds": row["seconds"],
    }


def tune(model_key):
    """Select the configuration of `model_key` with the lowest (best) validation MAE.

    Boosters and the linear model:
    - SEARCH over the walk-forward folds
    - then one run_split with the winner to build it's forecast for the validation year
    - builds forecast for the rolling interval band later (95% CI)
    """
    if MODELS[model_key]["family"] == "sarimax":
        return tune_sarimax(model_key)
    started = time.perf_counter()
    configs, table = search(model_key)  # get all paramters from grid-/randomized search
    best = table["MAE"].idxmin() if table["MAE"].notna().any() else None # get best found paramters + model config
    selected = configs[best] if best is not None else None # winning full parameter dict
    forecast, log = None, None
    if selected is not None:
        forecast, log = run_split(model_key, selected, "validation_rolling") # produce winners validation forecast, fit winner again on same rolling schedule
    return {
        "table": table,
        "selected": selected,
        "forecast": forecast,
        "log": log,
        "seconds": time.perf_counter() - started,
    }


def evaluate_model(model_key):
    """
    Entry point for one model:
    - tunes once
    - freezes winner
    - forecasts every enabled run (static / rolling) with it
    """
    tuning = tune(model_key)
    result = {"tuning": tuning, "runs": {}}
    if tuning["selected"] is None:
        return result
    result["runs"]["validation_rolling"] = (tuning["forecast"], tuning["log"])
    for run in ["validation_static", "test_static", "test_rolling"]:
        if RUNS[run]["enabled"]:
            result["runs"][run] = run_split(model_key, tuning["selected"], run)
    return result
````

### <a id="sec-4-1"></a>4.1 Features

`build_features` builds every feature group into `FEATURE_TABLE`: one row per target hour, built
**as at its own issue time**. The booster pipelines ([§4.5](#sec-4-5)) only select the enabled columns from it.

| Group | Columns | Available because |
|---|---|---|
| calendar | local hour, day of week, month, weekend, holiday | from the timestamp alone |
| smard_forecast_renewables | `fc_gen_wind_solar` for the target hour | published by 18:00 on `DAY−1` |
| lags | `grid_load` at the same local hour on `DAY−2` and `DAY−7`; the last value before the cutoff | observed before the cutoff ([§2.2](#sec-2-2)) |
| recent_smard_error | mean `err_renewables` over the 24 h ending at the cutoff (no `err_grid_load`: it holds the benchmark) | observed before the cutoff |
| capacity | `cap_wind_off + cap_wind_on + cap_solar` as published by the issue time | publication rule ([§2.3](#sec-2-3)) |
| cyclical | sine and cosine of the local time of day, the weekday and the day of year (over 365.25 days); `season_code` (winter 0 .. autumn 3) | from the timestamp alone |
| renewables_history | wind + solar capacity factor (`renewables / cap_total` at that hour) and SMARD's wind + solar error, both at the same local hour on `DAY−2` | observed before the cutoff ([§2.2](#sec-2-2)) |
| rolling_stats | mean of `grid_load` over the 72 h and standard deviation over the 168 h ending at the cutoff | observed before the cutoff |

- **Built before the pipeline, not in it:** lags, recent errors, capacity and the history and rolling
  columns depend on the cutoff and on other rows of the record, and a pipeline step sees only the rows it is given. No feature learns
  from the training data, so building them once is leakage-safe; [§5.4](#sec-5-4) tests it.
- Lags are looked up **by timestamp** (`hour − lag`), never by row position: the record skips the
  spring-DST hour, so a row shift would pick the wrong hour.
- A lag whose source hour does not exist (local 02:00 on a spring DST day) stays empty in
  `FEATURE_TABLE`. Both boosters handle missing values natively; the linear model fills them
  inside its pipeline with the training window's median ([§4.5](#sec-4-5)). No row is dropped.
- The rolling windows need every hour of the window: a window that contains a spring-DST switch is
  one hour short, so `gl_roll_mean_72h` stays empty for 3 days and `gl_roll_std_168h` for 7 days
  after each switch. They are filled the same way as the lags.
- **The three groups come from spec 06.1** (cutoff-safe versions of spec 05's features), with
  different column names than [`regression-models-magc.ipynb`](regression-models-magc.ipynb):
  `capacity_factor_lag_48h` is `vre_cf_lag_48h` there, and `renewables_history` +
  `rolling_stats` are its single group `fe_history`. `calendar` and `cyclical` are independent
  switches; `is_weekend` and `holiday` belong to `calendar` only.
- `build_features` takes the data as arguments, so the leakage test ([§5.4](#sec-5-4)) can rebuild the rows
  from data cut at a day's cutoff.
- `build_features` builds every group, whatever `USE_FEATURE` says; the pipeline's `select` step
  keeps only the enabled columns ([§4.5](#sec-4-5)).

**Adding a feature group**

1. [§1.3](#sec-1-3): a switch in `USE_FEATURE` and its column names in `FEATURE_GROUPS`, both at the end
   (template lines are there).
2. [§4.1](#sec-4-1): the columns at the end of `build_features`, in the same order (the [§4.6](#sec-4-6) self-check
   compares the order).
3. A feature built from a raw measured column that is new to the record: add that column to
   `MEASURED_COLUMNS` ([§5.4](#sec-5-4)), or the leakage test cannot mask it.

````python
DE_HOLIDAYS = holidays.country_holidays("DE", years=YEARS)  # only selects country holidays - no "Bundesland"


def is_holiday(index):
    """1.0 where the local date of a timestamp is a German federal holiday, else 0.0."""
    return pd.Series(index.date, index=index).isin(set(DE_HOLIDAYS)).astype(float)


def build_features(days, frame=time_series, errors=smard_errors):
    """Feature columns for every hour of `days`, each built as at its issue time.

    The leakage test passes `frame` and `errors` cut at a cutoff.
    """
    hours = frame.index[frame.index.normalize().isin(days)]
    setting = ROW_FORECAST_SETTING.loc[hours]
    last_available = pd.DatetimeIndex(setting["cutoff"] - RESOLUTION)
    actual = frame[TARGET]
    rows = pd.DataFrame(index=hours)

    # calendar: from the timestamps alone
    rows["hour"] = hours.hour
    rows["dow"] = hours.dayofweek
    rows["month"] = hours.month
    rows["is_weekend"] = (hours.dayofweek >= 5).astype(float)
    rows["holiday"] = is_holiday(hours).to_numpy()

    # smard_forecast
    rows["fc_gen_wind_solar"] = frame.loc[hours, "fc_gen_wind_solar"].to_numpy()

    # lags: by timestamp, never by row position
    for lag, name in LAG_NAME.items():
        rows[name] = actual.reindex(hours - lag).to_numpy()
    rows["gl_last_available"] = actual.reindex(last_available).to_numpy()

    # recent_smard_error
    error_columns = ["err_renewables"]
    window = FEATURE_WINDOWS["recent_smard_error"]
    recent = errors[error_columns].rolling(window).mean().reindex(last_available)
    rows["err_renewables_recent"] = recent["err_renewables"].to_numpy()

    # capacity
    capacity = published_capacity(setting["issue_time"], capacity_table(frame))
    rows["cap_total"] = capacity["cap_total"].to_numpy()

    # cyclical: sine / cosine of the local clock, weekday and day of year
    time_of_day = (hours - hours.normalize()) / pd.Timedelta(days=1)  # 0 .. <1
    days_per_year = FEATURE_WINDOWS["year_period"] / pd.Timedelta(days=1)
    year_share = hours.dayofyear / days_per_year
    rows["hour_sin"] = np.sin(2 * np.pi * time_of_day)
    rows["hour_cos"] = np.cos(2 * np.pi * time_of_day)
    rows["dow_sin"] = np.sin(2 * np.pi * hours.dayofweek / 7)
    rows["dow_cos"] = np.cos(2 * np.pi * hours.dayofweek / 7)
    rows["doy_sin"] = np.sin(2 * np.pi * year_share)
    rows["doy_cos"] = np.cos(2 * np.pi * year_share)
    season = hours.month.map(SEASON_OF_MONTH)
    rows["season_code"] = season.map(SEASON_ORDER.index).astype(int)  # winter = 0

    # renewables_history: at the same local hour on DAY−2, with that hour's capacity
    history_hours = hours - HISTORY_LAG
    cap_sum = frame[CAP_COLUMNS].sum(axis=1, min_count=len(CAP_COLUMNS))
    capacity_factor = (frame["renewables"] / cap_sum).reindex(history_hours)
    history_error = errors["err_renewables"].reindex(history_hours)
    rows[HISTORY_NAME["capacity_factor"]] = capacity_factor.to_numpy()
    rows[HISTORY_NAME["err_renewables"]] = history_error.to_numpy()

    # rolling_stats: TARGET over the windows ending at the cutoff
    mean_window = FEATURE_WINDOWS["roll_mean"]
    std_window = FEATURE_WINDOWS["roll_std"]
    rolling_mean = actual.rolling(mean_window, min_periods=n_obs(mean_window)).mean()
    rolling_std = actual.rolling(std_window, min_periods=n_obs(std_window)).std()
    rows[ROLLING_NAME["roll_mean"]] = rolling_mean.reindex(last_available).to_numpy()
    rows[ROLLING_NAME["roll_std"]] = rolling_std.reindex(last_available).to_numpy()

    # New group: its columns here, in FEATURE_GROUPS order; its switch in USE_FEATURE
    # rows["my_column"] = ...
    return rows


FEATURE_TABLE = build_features(DAYS)
group_columns = [column for columns in FEATURE_GROUPS.values() for column in columns]

# Empty values from the earliest training day of any fit on: what the models see
first_training_day = min(training_days(run["fit_days"][0])[0] for run in RUNS.values())
missing = FEATURE_TABLE.loc[first_training_day:].isna().sum()
print(
    f"\nFEATURE_TABLE: {len(FEATURE_TABLE):,} rows x {FEATURE_TABLE.shape[1]} columns. "
    f"Empty values from the first training day ({first_training_day:%Y-%m-%d}) on:"
)
print(missing[missing > 0].to_string() if missing.any() else "  none")

# Three target hours of the first test day: first, middle, last
first_test_hours = FORECAST_SETTING.at[TEST_DAYS[0], "target_hours"]
example_hours = first_test_hours[[0, len(first_test_hours) // 2, -1]]
print(
    "\nModel input of three target hours of the first test day "
    f"({TEST_DAYS[0]:%Y-%m-%d}; MWh, capacity in MW)"
)
example_input = FEATURE_TABLE.loc[example_hours]
display(example_input.T.rename(columns=lambda t: f"{t:%H:%M}"))
````

### <a id="sec-4-2"></a>4.2 Seasonal naive `DAY−7`

The floor: the actual `grid_load` at the same local hour seven days earlier. It needs no fit and
is identical in both split methods, so it appears once, with `split_method = "none"`. If `DAY−7` is a

### 4.3 `sarimax_fourier`

A regression with ARIMA errors, everything fixed in its registry entry ([§1.3](#sec-1-3)):

- **Exogenous inputs:** the `FEATURE_TABLE` columns named in `exog`, and Fourier terms for the
  periods and orders in `fourier`, computed on the local clock. Any feature from [§4.1](#sec-4-1) can be named
  in `exog`.
- **Error model:** ARIMA with the fixed `order` and `trend`, no order search. With `d = 0` the model
  reads as "regression on the inputs plus seasons, with AR/MA errors that carry the recent error
  forward". With `d = 1` it would carry the level of the last actual forward over the 9–32 h to the
  target hours (from the last usable actual, the row stamped 15:00 on `DAY−1`).
- **Fit** on the training days' rows. **Forecast:** the fitted parameters are applied once (`apply`,
  no re-estimation) to the series from the training window's start to the end of the fit's forecast
  stretch. Each day is then one dynamic prediction, from the first row not yet available (the row
  stamped 16:00 on `DAY−1`) to the end of `DAY`; only `DAY`'s hours are kept. A dynamic prediction
  uses only observations before its start, so no daily state updates are needed; the leakage test
  ([§5.4](#sec-5-4)) confirms it.
- **Series:** the local-time rows as evenly spaced steps: no UTC grid and no filling. The missing
  spring 02:00 is one skipped step per year; the autumn fold, which SMARD collapses into one row,
  needs nothing. The spacing is therefore off by one hour twice a year, which is negligible for
  hourly load. Rows with an empty input (an unforecastable day) are dropped the same way.

````python
def sarimax_exog(index, params):
    """SARIMAX's inputs for the rows `index`: its `exog` columns plus Fourier terms."""
    columns = {
        name: FEATURE_TABLE.loc[index, name].to_numpy() for name in params["exog"]
    }
    origin = pd.Timestamp(0)  # any fixed origin: sine and cosine absorb the phase
    for period, order in params["fourier"].items():
        phase = np.asarray(((index - origin) % period) / period)
        label = f"{period / pd.Timedelta(hours=1):g}h"
        for k in range(1, order + 1):
            columns[f"sin_{label}_{k}"] = np.sin(2 * np.pi * k * phase)
            columns[f"cos_{label}_{k}"] = np.cos(2 * np.pi * k * phase)
    return pd.DataFrame(columns, index=index)


def sarimax_series(first_row, last_row, params):
    """Rows from `first_row` to `last_row` with every input, and those inputs."""
    rows = time_series.loc[first_row:last_row].index
    exog = sarimax_exog(rows, params)
    keep = exog.notna().all(axis=1).to_numpy()
    return rows[keep], exog[keep]


def fit_sarimax(model_key, config, train_days, forecast_days):
    """Fit on the training days; forecast each day by one dynamic prediction.

    Each prediction starts at the day's first unavailable row.
    """
    train = hours_of(train_days)
    fitted = SARIMAX(
        time_series.loc[train, TARGET].to_numpy(),
        exog=sarimax_exog(train, config).to_numpy(),
        order=config["order"],
        trend=config["trend"],
    ).fit(disp=False)
    if not len(forecast_days):
        return lambda day: None

    rows, exog = sarimax_series(train[0], hours_of(forecast_days)[-1], config)
    endog = time_series.loc[rows, TARGET].to_numpy()
    applied = fitted.apply(endog=endog, exog=exog.to_numpy())

    def predict(day):
        hours = FORECAST_SETTING.at[day, "target_hours"]
        last_available = FORECAST_SETTING.at[day, "cutoff"] - RESOLUTION
        start = rows.searchsorted(last_available, side="right")  # first unavailable row
        end = rows.get_loc(hours[-1])
        mean = applied.get_prediction(start=start, end=end, dynamic=True).predicted_mean
        return pd.Series(mean, index=rows[start : end + 1])[hours]

    return predict


# SARIMAX's series in the default windows: rows dropped for an empty input, DST steps
SARIMAX_PARAMS = MODELS["sarimax_fourier"]["params"]
span_rows = time_series.loc[first_training_day : hours_of(TEST_DAYS)[-1]].index
kept_rows, _ = sarimax_series(span_rows[0], span_rows[-1], SARIMAX_PARAMS)
dropped_rows = len(span_rows) - len(kept_rows)
dst_steps = int(time_series.loc[span_rows, "spans_gap"].sum())
input_count = sarimax_exog(span_rows[:1], SARIMAX_PARAMS).shape[1]
print(
    f"SARIMAX series in the default windows: {first_training_day:%Y-%m-%d} .. "
    f"{TEST_END:%Y-%m-%d}, {len(span_rows):,} rows"
)
print(f"  rows dropped for a missing SMARD forecast : {dropped_rows}")
print(f"  spring-DST hours skipped as a single step : {dst_steps}")
print(f"  exogenous inputs                          : {input_count} plus the intercept")
````

### <a id="sec-4-4"></a>4.4 Stationarity check (KPSS)

A **check, not a decision**: one KPSS test on the residuals of an OLS fit of grid load on
`sarimax_fourier`'s exogenous inputs, over the first validation fit's training window (the
`WINDOWS["train"]` before the validation start). `d` stays at 0 whatever the test says; if KPSS rejects
stationarity, the team can change `d` in the registry.

- Differencing, if ever needed, happens **inside** ARIMA through `d`, never on the series outside the
  model.
- The boosters need no stationarity, but they **cannot extrapolate** beyond the target range they
  were trained on.
- Box-Cox or log transforms are not used. Grid load is always positive, so they would be possible
  here, unlike for residual load; they stay out to keep this notebook comparable with
  `regression-models-claude.ipynb`.

````python
# OLS on SARIMAX's inputs over the first validation fit's training window
kpss_rows = hours_of(training_days(RUNS["validation_rolling"]["fit_days"][0]))
ols = sm.OLS(
    time_series.loc[kpss_rows, TARGET].to_numpy(),
    sm.add_constant(sarimax_exog(kpss_rows, SARIMAX_PARAMS).to_numpy()),
).fit()

# KPSS on its residuals; a statistic outside KPSS's table gives the bound as p
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    kpss_stat, kpss_p, kpss_lags, kpss_crit = kpss(
        ols.resid, regression="c", nlags="auto"
    )
p_bounded = any(w.category.__name__ == "InterpolationWarning" for w in caught)
p_note = "outside the table (reported as the bound) " if p_bounded else ""
verdict = "rejects" if kpss_p < 0.05 else "does not reject"

print(
    f"KPSS on OLS residuals, {kpss_rows[0]:%Y-%m-%d} .. {kpss_rows[-1]:%Y-%m-%d} "
    f"({len(kpss_rows):,} h)"
)
print(
    f"  statistic {kpss_stat:.3f} with {kpss_lags} lags; "
    f"5 % critical value {kpss_crit['5%']}"
)
print(f"  p-value {p_note}{kpss_p:.3f}")
print(
    f"  -> {verdict} stationarity at 5 %. "
    f"d stays {SARIMAX_PARAMS['order'][1]} (registry); this is a check, not a decision."
)
````

### <a id="sec-4-5"></a>4.5 Boosters (direct and hybrid) and the linear model

LightGBM and XGBoost with the fixed seed `SEED`, each in two architectures, and a Ridge regression,
all built as sklearn pipelines by `build_pipeline`:

- **Direct** (`select → booster`): the booster predicts `grid_load` from the enabled feature
  groups.
- **Hybrid** (`HybridRegressor`): the **linear stage** is an OLS on grid load at the same
  local hour on `DAY−2` and `DAY−7` (`gl_lag_48h`, `gl_lag_168h`) and a linear trend (days since the
  first training day). A lag whose source hour does not exist (the spring-DST 02:00) is filled with
  the training window's median inside the stage, because OLS takes no empty value. The **residual
  stage** (`select → booster`) is fitted on the linear stage's in-sample residuals; the forecast is
  linear + residual stage. **The linear stage's inputs are fixed** (`LINEAR_INPUTS`): it reads
  them from the full row before `select`, so the `USE_FEATURE` switches affect only the booster.
- **Linear** (`select → prepare → Ridge`, `linear_direct`): the same enabled feature groups as the
  direct boosters. `prepare` one-hot encodes `hour`, `dow` and `month` (`CALENDAR_CATEGORIES`),
  fills empty values with the median and scales the other columns. It is fitted inside the
  pipeline, so the medians and scales come from each training window only. Families that need
  this step are listed in `PREPROCESS`; Ridge sits in the step named `booster`, so its grid
  parameters reach it as `booster__<name>`, like a direct booster's.
- `select` passes the enabled `USE_FEATURE` columns through unchanged. A grid parameter reaches the
  booster as `booster__<name>` (direct) or `residual__booster__<name>` (hybrid), recorded in
  `PARAM_PREFIX`.
- **No early stopping.** `n_estimators` is a grid value; early stopping would need a third
  time-ordered split inside each training window.

**Expected difference:** the hybrid can extrapolate the level through its linear stage, while the direct
booster is capped at the range of its training targets. This matters in the tails, where the test
year may reach a higher or lower load than the training window; the extremes scoreboard ([§7.3](#sec-7-3))
shows it as a bias in `low_extreme` / `high_extreme` and as few forecast hours in those bins. The linear
model is not capped either: it extrapolates through every input.

````python
# fixed: independent of the USE_FEATURE switches (the same-hour lags of TARGET)
LINEAR_INPUTS = list(LAG_NAME.values())
BOOSTER = {
    "lightgbm": lgb.LGBMRegressor,
    "xgboost": xgb.XGBRegressor,
    "linear": Ridge,  # no booster, but built and tuned the same way
}

# Families that need a prepare step: no NaN, calendar one-hot encoded
PREPROCESS = {"linear"}
CALENDAR_CATEGORIES = ["hour", "dow", "month"]

# Booster pipeline prefixes
PARAM_PREFIX = {"direct": "booster__", "hybrid": "residual__booster__"}


def feature_columns():
    """The model-input columns of every enabled USE_FEATURE group."""
    enabled_groups = [group for group, enabled in USE_FEATURE.items() if enabled]
    return [column for group in enabled_groups for column in FEATURE_GROUPS[group]]


def select(columns):
    """Pipeline step keeping `columns` in that order, as a DataFrame."""
    return ColumnTransformer(
        [("features", "passthrough", columns)],
        remainder="drop",
        verbose_feature_names_out=False,
    ).set_output(transform="pandas")


def prepare(columns):
    """Pipeline step for a linear model: one-hot calendar, impute + scale the rest."""
    categorical = [column for column in CALENDAR_CATEGORIES if column in columns]
    numeric = [column for column in columns if column not in categorical]
    one_hot = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    impute_and_scale = Pipeline(
        [("impute", SimpleImputer(strategy="median")), ("scale", StandardScaler())]
    )
    return ColumnTransformer(
        [("calendar", one_hot, categorical), ("numeric", impute_and_scale, numeric)]
    )


class HybridRegressor(RegressorMixin, BaseEstimator):
    """
    Linear stage: OLS on `linear_inputs` and a linear trend; empty inputs filled with
    the training median (the lag of a spring-DST hour has no source row).
    Residual stage: `residual` fitted on the linear stage's in-sample errors.

    The trend counts days since the first training day, learned at fit.
    The forecast is linear stage + residual stage.
    """

    def __init__(self, residual=None, linear_inputs=tuple(LINEAR_INPUTS)):
        self.residual = residual
        self.linear_inputs = linear_inputs

    def _linear_design(self, X):
        trend = np.asarray((X.index - self.origin_) / pd.Timedelta(days=1))
        return np.column_stack([X[list(self.linear_inputs)].to_numpy(), trend])

    def fit(self, X, y):
        self.origin_ = X.index.min().normalize()
        design = self._linear_design(X)
        self.linear_ = Pipeline(
            [("impute", SimpleImputer(strategy="median")), ("ols", LinearRegression())]
        ).fit(design, y)
        linear_error = np.asarray(y) - self.linear_.predict(design)
        self.residual_ = clone(self.residual).fit(X, linear_error)
        return self

    def predict(self, X):
        return self.linear_.predict(self._linear_design(X)) + self.residual_.predict(X)


def build_pipeline(model_key, config):
    """The unfitted pipeline of booster `model_key`, with `config` on its booster.

    direct: select -> booster (PREPROCESS families: select -> prepare -> booster)
    hybrid: HybridRegressor(linear stage, residual stage = select -> booster)
    """
    columns = feature_columns()
    if not columns:
        raise ValueError("every USE_FEATURE group is switched off")

    family = MODELS[model_key]["family"]
    estimator = BOOSTER[family](**config)
    steps = [("select", select(columns))]
    if family in PREPROCESS:
        steps.append(("prepare", prepare(columns)))
    booster = Pipeline(steps + [("booster", estimator)])
    if MODELS[model_key]["architecture"] == "hybrid":
        return HybridRegressor(residual=booster)
    return booster


def fit_booster(model_key, config, train_days, forecast_days):
    """Fit `model_key` on the training days; forecast all `forecast_days` at once."""
    train = FEATURE_TABLE.loc[hours_of(train_days)]
    target = time_series.loc[train.index, TARGET]
    pipeline = build_pipeline(model_key, config).fit(train, target)
    if not len(forecast_days):
        return lambda day: None

    future = FEATURE_TABLE.loc[hours_of(forecast_days)]
    forecast = pd.Series(pipeline.predict(future), index=future.index)
    return lambda day: forecast[FORECAST_SETTING.at[day, "target_hours"]]


FAMILY_FIT.update(
    {
        "sarimax": fit_sarimax,
        "lightgbm": fit_booster,
        "xgboost": fit_booster,
        "linear": fit_booster,
    }
)

features_in_use = feature_columns()
linear_text = ", ".join(LINEAR_INPUTS)
print(f"booster features in use ({len(features_in_use)}): {', '.join(features_in_use)}")
print(f"hybrid linear stage inputs (fixed): {linear_text}, trend in days, intercept; empty lags: training median")

# One enabled booster pipeline per architecture
for architecture in ["direct", "hybrid"]:
    example_keys = [
        key
        for key, model in MODELS.items()
        if model["enabled"]
        and model["architecture"] == architecture
        and model["family"] not in PREPROCESS
    ]
    if example_keys:
        example_key = example_keys[0]
        print(
            f"\n{architecture} pipeline, e.g. {example_key} "
            f"(booster parameters under '{PARAM_PREFIX[architecture]}'):"
        )
        display(build_pipeline(example_key, MODELS[example_key]["params"]))

# The linear pipeline, with its prepare step
linear_keys = [
    key
    for key, model in MODELS.items()
    if model["enabled"] and model["family"] in PREPROCESS
]
if linear_keys:
    print(f"\nlinear pipeline, e.g. {linear_keys[0]} (parameters under 'booster__'):")
    display(build_pipeline(linear_keys[0], MODELS[linear_keys[0]]["params"]))
````

### <a id="sec-5-4"></a>5.4 Leakage test by truncation

For a fixed-seed sample of delivery days from the training, validation and test windows, plus the
two days where leakage is easiest to get wrong (the latest 1 January and the latest spring-DST day),
the data is **cut off at that day's availability cutoff**: actuals and SMARD's errors up to the
cutoff, SMARD's forecasts up to the end of `DAY`.

- The day's feature rows are rebuilt from the cut data and must be **identical** to the rows the
  models used. Any feature that peeks past the cutoff changes when the future is removed, so this
  catches leakage without tracking a timestamp per feature.
- When `sarimax_fourier` is switched on: for the sampled validation and test days, its forecast from
  the fitting run must equal a dynamic prediction from the same fit applied to the series cut at the
  cutoff.
- Four fitting rules are asserted:
  - tuning and every validation fit never train on a test day; the KPSS OLS ([§4.4](#sec-4-4)) sees no
    validation or test row
  - the test year is forecast but never used for selection: tuning forecasts validation hours only
  - every fit trains only on days before its fit day, and the fits run in time order (no shuffling
    anywhere)
  - every search fold trains only on hours before those it scores and scores validation hours only

````python
LEAKAGE_SAMPLE_DAYS = 20        # drawn with SEED, spread evenly over the three windows
MEASURED_COLUMNS = [
    "wind_off",
    "wind_on",
    "solar",
    "grid_load",
    "residual_load",
    "renewables",
    *CAP_COLUMNS,
]


def cut_at_cutoff(day):
    """The data as known at `day`'s issue time.

    Measured values up to the cutoff, SMARD forecasts to the end of `day`.
    """
    cutoff = FORECAST_SETTING.at[day, "cutoff"]
    end = FORECAST_SETTING.at[day, "target_hours"][-1]
    frame = time_series.loc[:end].copy()
    errors = smard_errors.loc[:end].copy()
    frame.loc[~available(frame.index, cutoff), MEASURED_COLUMNS] = np.nan
    # every error needs the actual
    errors.loc[~available(errors.index, cutoff), :] = np.nan
    return frame, errors


# Sample days per window, spread evenly and drawn with SEED
pools = {
    "training": training_days(RUNS["validation_rolling"]["fit_days"][0]),
    "validation": VAL_DAYS[FORECASTABLE.loc[VAL_DAYS].to_numpy()],
    "test": TEST_DAYS[FORECASTABLE.loc[TEST_DAYS].to_numpy()],
}
rng = np.random.default_rng(SEED)
bounds = np.linspace(0, LEAKAGE_SAMPLE_DAYS, len(pools) + 1).round().astype(int)
per_pool = np.diff(bounds)
SAMPLE = {
    name: pd.DatetimeIndex(sorted(rng.choice(days, size=n, replace=False)))
    for (name, days), n in zip(pools.items(), per_pool)
}

# Add the edge days: latest 1 January and latest spring-DST day
for edge in [new_year, dst_days[-1]]:
    for name, days in pools.items():
        if edge in days and edge not in SAMPLE[name]:
            SAMPLE[name] = SAMPLE[name].append(pd.DatetimeIndex([edge])).sort_values()

# 1. Feature rows rebuilt from the cut data
feature_checks = []
for window, days in SAMPLE.items():
    for day in days:
        frame, errors = cut_at_cutoff(day)
        rebuilt = build_features([day], frame, errors)
        identical = rebuilt.equals(FEATURE_TABLE.loc[rebuilt.index])
        feature_checks.append({"window": window, "day": day, "identical": identical})
feature_checks = pd.DataFrame(feature_checks)

changed_days = feature_checks.loc[~feature_checks["identical"], "day"].tolist()
message = f"feature rows changed under truncation: {changed_days}"
assert feature_checks["identical"].all(), message

print(
    f"Leakage test by truncation: {len(feature_checks)} delivery days (seed {SEED}, "
    f"incl. {new_year:%Y-%m-%d} and {dst_days[-1]:%Y-%m-%d})"
)
````

````python
SARIMAX_TOLERANCE = 1e-6        # MWh: floating-point identical

# 2. SARIMAX: the fitting run's forecast against the same fit applied to the cut series
sarimax_checks = []
if RESULTS.get("sarimax_fourier", {}).get("runs"):
    config = SELECTED["sarimax_fourier"]
    fits = {}
    window_runs = [
        ("validation", ["validation_static", "validation_rolling"]),
        ("test", ["test_static", "test_rolling"]),
    ]
    for window, candidates in window_runs:
        run = next(r for r in candidates if r in RESULTS["sarimax_fourier"]["runs"])
        production = RESULTS["sarimax_fourier"]["runs"][run][0]
        fit_days = RUNS[run]["fit_days"]
        for day in SAMPLE[window]:
            fit_day = fit_days[fit_days.searchsorted(day, side="right") - 1]

            # Refit once per fit day, as in the fitting run
            if fit_day not in fits:
                train = hours_of(training_days(fit_day))
                with warnings.catch_warnings():
                    warnings.simplefilter("ignore")
                    sarimax = SARIMAX(
                        time_series.loc[train, TARGET].to_numpy(),
                        exog=sarimax_exog(train, config).to_numpy(),
                        order=config["order"],
                        trend=config["trend"],
                    )
                    fits[fit_day] = (train[0], sarimax.fit(disp=False))
            first_row, fitted = fits[fit_day]

            # Apply the fit to the series cut at the cutoff, predict dynamically
            hours = FORECAST_SETTING.at[day, "target_hours"]
            cutoff = FORECAST_SETTING.at[day, "cutoff"]
            rows, exog = sarimax_series(first_row, hours[-1], config)
            known = available(rows, cutoff)
            endog = time_series.loc[rows, TARGET].where(known).to_numpy()
            applied = fitted.apply(endog=endog, exog=exog.to_numpy())
            start = rows.searchsorted(cutoff - RESOLUTION, side="right")
            end = rows.get_loc(hours[-1])
            prediction = applied.get_prediction(start=start, end=end, dynamic=True)
            cut = pd.Series(prediction.predicted_mean, index=rows[start : end + 1])

            difference = float((cut[hours] - production[hours]).abs().max())
            sarimax_checks.append(
                {
                    "window": window,
                    "day": day,
                    "run": run,
                    "max_abs_difference": difference,
                }
            )
sarimax_checks = pd.DataFrame(sarimax_checks)

within_tolerance = (
    sarimax_checks.empty
    or (sarimax_checks["max_abs_difference"] <= SARIMAX_TOLERANCE).all()
)
assert within_tolerance, sarimax_checks
````

````python
# The search folds (§3.3) never reach FIT_LOG: check them directly
SEARCH_FOLDS = walk_forward_folds("validation_rolling")
for train, test in SEARCH_FOLDS:
    fold_train, fold_test = FEATURE_TABLE.index[train], FEATURE_TABLE.index[test]
    trains_before = fold_train.max() < fold_test.min()
    message = "a search fold trains on an hour at or after the hours it scores"
    assert trains_before, message
    message = "a search fold scores an hour outside the validation year"
    assert fold_test.isin(hours_of(VAL_DAYS)).all(), message
    assert fold_train.max() < TEST_DAYS[0], "a search fold trains on a test day"

# 3. Fitting rules
for model_key, result in RESULTS.items():
    tuning = result["tuning"]
    only_validation = tuning["forecast"].index.equals(hours_of(VAL_DAYS))
    message = f"{model_key}: selection saw hours outside the validation year"
    assert only_validation, message
    assert (tuning["table"]["hour_count"] <= len(hours_of(VAL_DAYS))).all()
    for run, (forecast, log) in result["runs"].items():
        assert forecast.index.equals(hours_of(RUNS[run]["days"]))
        in_time_order = (
            log["fit_day"].is_monotonic_increasing and log["fit_day"].is_unique
        )
        assert in_time_order, f"{model_key}/{run}: fits out of time order"
        for fit_day in log["fit_day"]:
            train = training_days(fit_day)
            before_fit_day = train.is_monotonic_increasing and (train < fit_day).all()
            message = f"{model_key}/{run}: training after the fit day"
            assert before_fit_day, message
            if run.startswith("validation"):
                message = (
                    f"{model_key}/{run}: a tuning or calibration fit saw a test day"
                )
                assert (train < TEST_DAYS[0]).all(), message
message = "the KPSS OLS must not see validation or test rows"
assert kpss_rows[-1] < VAL_DAYS[0], message
LEAKAGE_PASSED = True

print(
    f"passed: no rebuilt feature changed, SARIMAX ignores everything "
    f"after the cutoff, {len(SEARCH_FOLDS)} search folds time-ordered, "
    f"all fitting rules hold"
)
````

## <a id="sec-6"></a>6 Prediction intervals

**Prediction interval calculation** for both split variants (and seasonal naive):

- **Residual sign:** the band uses `r = actual − forecast`, the **opposite** sign to the project's
  `error = forecast − actual`. A positive `r` means the actual came out above the forecast.
- **Calibration, per split method,** the way the method is used:
  - rolling: the residuals of the tuning walk-forward ([§3.3](#sec-3-3)) of the selected configuration
  - static: the residuals of one frozen fit at the start of the validation year, forecasting the
    whole validation year without refitting. A model frozen for a year makes larger errors than a
    refitted one, so rolling residuals would make the static band too narrow.
  - seasonal naive: its validation-year residuals, for its single row
- **Quantiles:** the `(1 − level) / 2` and `(1 + level) / 2` quantiles of `r` **per local hour of
  day**, because the error size varies strongly with the hour. Example: `level = 0.95`
  (`INTERVAL`, [§1.3](#sec-1-3)) gives the 0.025 and 0.975 quantiles.
- **Band:** `forecast + [q_low(r), q_high(r)]` for each test hour, with the quantiles of that hour.

The band is calibrated on the validation year and **not updated** during the test year, so drift
shows up as lost coverage (scoreboard C, [§7.4](#sec-7-4)). If a model is strongly biased at some hour, both
quantiles share a sign and the band does not contain the forecast itself; that is correct, not an
error. SMARD is a point forecast and gets no band.

````python
LOW_Q, HIGH_Q = (1 - INTERVAL["level"]) / 2, (1 + INTERVAL["level"]) / 2
CALIBRATED_ON = {
    "test_static": "validation_static",
    "test_rolling": "validation_rolling",
}


def band_quantiles(forecast):
    """Per local hour: the LOW_Q and HIGH_Q quantiles of r = actual − forecast."""
    r = (time_series.loc[forecast.index, TARGET] - forecast).dropna()
    quantiles = r.groupby(r.index.hour).quantile([LOW_Q, HIGH_Q]).unstack()
    return quantiles.set_axis(["q_low", "q_high"], axis=1)


def with_band(forecast, quantiles):
    """The forecast plus its band, using the quantiles of each hour's local hour."""
    q = quantiles.reindex(forecast.index.hour).set_axis(forecast.index)
    return pd.DataFrame(
        {
            "forecast": forecast,
            "lower": forecast + q["q_low"],
            "upper": forecast + q["q_high"],
        }
    )


# (model, split_method) -> test-hour forecast with band; CALIBRATION: for the self-check
FORECASTS, CALIBRATION = {}, {}
for model_key, result in RESULTS.items():
    for run, calibration_run in CALIBRATED_ON.items():
        if run in result["runs"] and calibration_run in result["runs"]:
            row = (model_key, RUNS[run]["split_method"])
            quantiles = band_quantiles(result["runs"][calibration_run][0])
            CALIBRATION[row] = (calibration_run, quantiles)
            FORECASTS[row] = with_band(result["runs"][run][0], quantiles)

# Seasonal naive: calibrated on its validation-year forecast
naive_quantiles = band_quantiles(NAIVE["validation"])
CALIBRATION[("seasonal_naive", "none")] = ("validation year", naive_quantiles)
FORECASTS[("seasonal_naive", "none")] = with_band(NAIVE["test"], naive_quantiles)
````

## 7 Evaluation and scoreboard

**Rows:** every enabled registry model × split method with at least one forecast, seasonal naive
(`split_method = "none"`) and SMARD's grid-load forecast `fc_grid_load` (`smard`, `"none"`).

**Common hours:** every row is scored on the same hours: the test hours where the actual, SMARD's
forecast, seasonal naive's forecast and every registry row's forecast all exist (a strict
intersection). SMARD is re-scored from `smard_forecast_errors_hourly.csv` on these hours, and a
cross-check confirms the file agrees with `data/smard.csv`. A row with no forecast at all (a failed
static fit) gets no scoreboard row and is left out of the intersection, so it cannot empty everyone
else's evaluation set. If one model breaks badly, switch it off in the registry and re-run.

**Metrics** follow spec 04:

- `error = forecast − actual`; **MAE**, **RMSE** (recomputed from the hourly errors, never averaged)
  and **bias**; no MAPE. The hour count is in the header line above each table and in the export's
  `count` column
- **skill vs SMARD** = `100 · (1 − MAE_model / MAE_SMARD)` on the same hours, in `%`; **positive =
  better than SMARD**
- **months beating SMARD**: in how many of the `n` calendar months that lie entirely inside the test
  window the row's MAE is below SMARD's (`n` is in the header line). This is the robustness view in
  place of a significance test.

### 7.1 Common hours

````python
actual = time_series[TARGET]
test_hours = hours_of(TEST_DAYS)
SMARD_ROW = ("smard", "none")

# Forecast per row: every row with at least one forecast, plus SMARD
ROWS = {
    row: frame["forecast"]
    for row, frame in FORECASTS.items()
    if frame["forecast"].notna().any()
}
ROWS[SMARD_ROW] = smard_errors.loc[test_hours, BENCHMARK]

# Common hours: test hours with an actual and a forecast from every row
has_actual = actual[test_hours].notna()
forecast_exists = pd.DataFrame(
    {row: forecast.notna() for row, forecast in ROWS.items()}, index=test_hours
)
COMMON = test_hours[(forecast_exists.all(axis=1) & has_actual).to_numpy()]

# SMARD re-scored from its hourly errors file, cross-checked against data/smard.csv
smard_error = smard_errors.loc[COMMON, BENCHMARK_ERROR]
csv_error = time_series.loc[COMMON, BENCHMARK] - actual[COMMON]
actual_difference = smard_errors.loc[COMMON, TARGET] - actual[COMMON]
SMARD_CROSS_CHECK = max(
    float((smard_error - csv_error).abs().max()),
    float(actual_difference.abs().max()),
)
assert SMARD_CROSS_CHECK <= 1e-6, (
    "smard_forecast_errors_hourly.csv disagrees with data/smard.csv "
    f"by {SMARD_CROSS_CHECK} MWh"
)

# Error per row on the common hours
ERRORS = {
    row: forecast[COMMON] - actual[COMMON]
    for row, forecast in ROWS.items()
    if row != SMARD_ROW
}
ERRORS[SMARD_ROW] = smard_error

lost_hours = len(test_hours) - len(COMMON)
print(f"test window {TEST_DAYS[0]:%Y-%m-%d} .. {TEST_DAYS[-1]:%Y-%m-%d}")
print(f"  hours in the test window             : {len(test_hours):,}")
print(f"  hours with an actual                 : {int(has_actual.sum()):,}")
print(f"  common hours (actual + all forecasts): {len(COMMON):,} ({lost_hours:,} lost)")
print(f"SMARD forecast_errors_hourly.csv vs data/smard.csv: max |difference| {SMARD_CROSS_CHECK:.1e} MWh")
````

### <a id="sec-7-2"></a>7.2 Scoreboard A: accuracy

Sorted by MAE. `fit_seconds` sums the split method's **test-year fits only**; [§5.1](#sec-5-1) prints each
model's total tuning and run time. SMARD has no skill, months or `fit_seconds` ("-"): the first two
compare a row against SMARD, and SMARD has no fit. Seasonal naive has no fit either.

**Why fewer than 12 months:** only calendar months that lie entirely inside the test window count.
The window is the last 365 days of the data and rarely starts on the 1st, so the months at both
ends are partial and left out. A window from 7 September to 6 September drops both partial
Septembers, which leaves 11. The header line shows the current count.

````python
SCOREBOARD_COLUMNS = ["model", "split_method", "table", "metric", "value", "count"]


def scoreboard_table(records):
    """One row per model and split method, one column per metric (values only)."""
    board = pd.DataFrame(records, columns=SCOREBOARD_COLUMNS)
    table = board.pivot(
        index=["model", "split_method"], columns="metric", values="value"
    )
    return table[board["metric"].unique()]


# Calendar months that lie entirely inside the test window
FULL_MONTHS = [
    month
    for month in TEST_DAYS.to_period("M").unique()
    if month.start_time in TEST_DAYS and month.end_time.normalize() in TEST_DAYS
]
month_of = COMMON.to_period("M")
monthly_mae = {
    row: error.abs().groupby(month_of).mean() for row, error in ERRORS.items()
}
MAE_SMARD = ERRORS[SMARD_ROW].abs().mean()
TEST_RUN = {RUNS[run]["split_method"]: run for run in ["test_static", "test_rolling"]}

# Metrics per row as (value, count)
ACCURACY_RECORDS = []
for row, error in ERRORS.items():
    hour_count, mae = len(error), error.abs().mean()
    metrics = {
        "MAE": (mae, hour_count),
        "RMSE": (np.sqrt((error**2).mean()), hour_count),
        "bias": (error.mean(), hour_count),
    }
    if row != SMARD_ROW:
        beats_smard = (
            monthly_mae[row][FULL_MONTHS] < monthly_mae[SMARD_ROW][FULL_MONTHS]
        )
        metrics["skill_pct"] = (100 * (1 - mae / MAE_SMARD), hour_count)
        metrics["months_beating_smard"] = (int(beats_smard.sum()), len(FULL_MONTHS))
    if row[0] in MODELS:
        in_run = (FIT_LOG["model"] == row[0]) & (FIT_LOG["run"] == TEST_RUN[row[1]])
        fit_seconds = FIT_LOG.loc[in_run, "fit_seconds"].sum()
        metrics["fit_seconds"] = (fit_seconds, int(in_run.sum()))
    for metric, (value, count) in metrics.items():
        ACCURACY_RECORDS.append((*row, "accuracy", metric, value, count))

SCOREBOARD_A = scoreboard_table(ACCURACY_RECORDS).sort_values("MAE")
accuracy_labels = {
    "skill_pct": "skill vs SMARD (%)",
    "months_beating_smard": "months beating SMARD",
}
accuracy_formats = {
    "MAE": "{:,.0f}",
    "RMSE": "{:,.0f}",
    "bias": "{:+,.0f}",
    "skill vs SMARD (%)": "{:+.1f}",
    "months beating SMARD": "{:.0f}",
    "fit_seconds": "{:,.1f}",
}
print(
    f"Scoreboard A: accuracy on {len(COMMON):,} common test hours "
    f"({TEST_DAYS[0]:%Y-%m-%d} .. {TEST_DAYS[-1]:%Y-%m-%d}); "
    f"MAE, RMSE and bias in MWh; {len(FULL_MONTHS)} full calendar months"
)
accuracy_table = SCOREBOARD_A.rename(columns=accuracy_labels)
display(accuracy_table.style.format(accuracy_formats, na_rep="-"))
````

### <a id="sec-7-3"></a>7.3 Scoreboard B: extremes

**Tail bins:** the edges are the `{1, 25, 75, 99} %` quantiles of the **actual** grid load over the
test window, computed once and shared by every row. MAE and bias are reported for `low_extreme`
(`≤ P1`), `ordinary` (`P25–P75`) and `high_extreme` (`> P99`), the two load extremes and the ordinary
hours between them. They are reported once **binned by the actual** and once **binned by the row's
own forecast**. Regression to the mean makes a forecast look biased towards the middle when binned
by the actual, and towards the tails when binned by the forecast, so only a tail difference that
shows in **both** views counts.

**Skill vs SMARD** is reported where a row and SMARD are scored on the same hours or days: binned by
the actual, and for the day maximum and minimum. Binned by the forecast, every row fills a bin with
its own hours (the `hours` columns), so a skill there would compare different hours.

**Day maximum and minimum:** the value error (`forecast − actual`) of each day's highest and lowest
hour, taken over the common hours only. A day counts if its common hours cover at least
`DAY_COMPLETENESS` of its expected hours.

````python
TAIL_LEVELS = [0.01, 0.25, 0.75, 0.99]
EDGES = actual[test_hours].quantile(TAIL_LEVELS)
P1, P25, P75, P99 = EDGES.to_numpy()
BINS = ["low_extreme", "ordinary", "high_extreme"]


def tail_bin(values):
    """low_extreme (<= P1), ordinary (P25..P75), high_extreme (> P99), else other."""
    conditions = [values <= P1, (values >= P25) & (values <= P75), values > P99]
    return pd.Series(np.select(conditions, BINS, default="other"), index=values.index)


# Days whose common hours cover at least DAY_COMPLETENESS of their expected hours
common_day = COMMON.normalize()
common_per_day = pd.Series(1, index=COMMON).groupby(common_day).sum()
expected_per_day = (
    FORECAST_SETTING["target_hours"].map(len).reindex(common_per_day.index)
)
is_complete = (common_per_day >= DAY_COMPLETENESS * expected_per_day).to_numpy()
DAYS_SCORED = common_per_day.index[is_complete]
scored = common_day.isin(DAYS_SCORED)


def day_value_error(row, agg):
    """forecast − actual of each scored day's maximum or minimum."""
    day = common_day[scored]
    forecast = ROWS[row][COMMON][scored].groupby(day).agg(agg)
    observed = actual[COMMON][scored].groupby(day).agg(agg)
    return forecast - observed


def group_metrics(error, smard_error=None):
    """MAE and bias of a group of errors, plus skill vs SMARD on the same group."""
    metrics = {"MAE": error.abs().mean(), "bias": error.mean()}
    if smard_error is not None:
        metrics["skill_pct"] = 100 * (1 - metrics["MAE"] / smard_error.abs().mean())
    return metrics


# One entry per row, view, bin and metric
extreme_stats = []
for row, error in ERRORS.items():
    for view, values in [("actual", actual[COMMON]), ("forecast", ROWS[row][COMMON])]:
        bins = tail_bin(values)
        for bin_name in BINS:
            in_bin = bins == bin_name
            with_skill = view == "actual" and row != SMARD_ROW
            reference = ERRORS[SMARD_ROW][in_bin] if with_skill else None
            for metric, value in group_metrics(error[in_bin], reference).items():
                extreme_stats.append(
                    (*row, view, bin_name, metric, value, in_bin.sum())
                )
    for agg in ["max", "min"]:
        day_error = day_value_error(row, agg)
        reference = day_value_error(SMARD_ROW, agg) if row != SMARD_ROW else None
        for metric, value in group_metrics(day_error, reference).items():
            extreme_stats.append((*row, "day", agg, metric, value, len(day_error)))

# Export names: {metric}_{bin}_by_{view} and {metric}_day_{max,min}
EXTREMES_RECORDS = []
for model, split, view, name, metric, value, count in extreme_stats:
    suffix = f"day_{name}" if view == "day" else f"{name}_by_{view}"
    EXTREMES_RECORDS.append(
        (model, split, "extremes", f"{metric}_{suffix}", value, count)
    )

# One table per view: bins as column groups, rows in Scoreboard A's order
extremes = pd.DataFrame(
    extreme_stats,
    columns=["model", "split_method", "view", "bin", "metric", "value", "count"],
)
bin_hours = tail_bin(actual[COMMON]).value_counts()
hours_text = ", ".join(f"{name} {bin_hours.get(name, 0):,}" for name in BINS)
titles = {
    "actual": f"error in the tails, binned by the actual (MWh); hours: {hours_text}",
    "forecast": "error in the tails, binned by the row's own forecast (MWh)",
    "day": f"value error of the day maximum and minimum (MWh), {len(DAYS_SCORED)} days",
}
extreme_formats = {
    "MAE": "{:,.0f}",
    "bias": "{:+,.0f}",
    "skill vs SMARD (%)": "{:+.1f}",
    "hours": "{:,.0f}",
}
edges_text = ", ".join(
    f"P{level * 100:g} = {edge:,.0f}" for level, edge in EDGES.items()
)
print("Tail-bin edges: quantiles of the actual over the test window (MWh)")
print(f"  {edges_text}")
for view, title in titles.items():
    in_view = extremes[extremes["view"] == view]
    table = in_view.pivot(
        index=["model", "split_method"], columns=["bin", "metric"], values="value"
    )
    if view == "forecast":
        is_mae = in_view["metric"] == "MAE"
        hours = in_view[is_mae].pivot(
            index=["model", "split_method"], columns="bin", values="count"
        )
        for name in BINS:
            table[(name, "hours")] = hours[name]
    table = table[in_view["bin"].unique()].loc[SCOREBOARD_A.index]
    table = table.rename(columns={"skill_pct": "skill vs SMARD (%)"}, level=1)
    table_formats = {column: extreme_formats[column[1]] for column in table.columns}
    print(f"\nScoreboard B: {title}")
    display(table.style.format(table_formats, na_rep="—"))
````

### <a id="sec-7-4"></a>7.4 Scoreboard C: intervals

Coverage is the share of common hours whose actual lies inside the row's band, against the nominal
level. A negative `coverage − nominal` means the band is too narrow. SMARD has no band and
therefore no row.

````python
# Coverage and mean width of each row's band; SMARD has no band
INTERVAL_RECORDS = []
observed = actual[COMMON]
for row, frame in FORECASTS.items():
    if row not in ROWS:
        continue
    band = frame.loc[COMMON]
    inside = (band["lower"] <= observed) & (observed <= band["upper"])
    coverage, width = 100 * inside.mean(), (band["upper"] - band["lower"]).mean()
    INTERVAL_RECORDS.append((*row, "intervals", "coverage_pct", coverage, len(COMMON)))
    INTERVAL_RECORDS.append((*row, "intervals", "mean_width", width, len(COMMON)))

interval_table = scoreboard_table(INTERVAL_RECORDS)
interval_table["coverage − nominal (pp)"] = (
    interval_table["coverage_pct"] - 100 * INTERVAL["level"]
)
interval_table = interval_table.rename(
    columns={"coverage_pct": "coverage (%)", "mean_width": "mean width (MWh)"}
)
row_order = [row for row in SCOREBOARD_A.index if row in interval_table.index]
interval_formats = {
    "coverage (%)": "{:.1f}",
    "mean width (MWh)": "{:,.0f}",
    "coverage − nominal (pp)": "{:+.1f}",
}
print(
    f"Scoreboard C: {INTERVAL['level']:.0%} prediction intervals "
    f"on {len(COMMON):,} common test hours"
)
display(interval_table.loc[row_order].style.format(interval_formats))
````

### <a id="sec-7-5"></a>7.5 Static vs. rolling

The value of refitting: each model's MAE difference `rolling − static` on the common hours. A negative
difference means refitting helped. Both split methods share their first fit, so the first
`refit_every` of the test year cannot differ between them ([§3.2](#sec-3-2)).

````python
# MAE per model and split method, from Scoreboard A
mae_by_split = SCOREBOARD_A["MAE"].unstack("split_method")
if {"static", "rolling"} <= set(mae_by_split.columns):
    refit = mae_by_split[["static", "rolling"]].dropna()
    refit["rolling − static"] = refit["rolling"] - refit["static"]
    refit["rolling − static (%)"] = 100 * (refit["rolling"] / refit["static"] - 1)
    refit = refit.rename(columns={"static": "MAE static", "rolling": "MAE rolling"})
    refit_formats = {
        "MAE static": "{:,.0f}",
        "MAE rolling": "{:,.0f}",
        "rolling − static": "{:+,.0f}",
        "rolling − static (%)": "{:+.1f}",
    }
    print(f"MAE rolling − static per model (MWh), {len(COMMON):,} common hours")
    display(refit.style.format(refit_formats))
else:
    print("both split methods are needed for this table")
````

### 7.6 Forecast plot with the band

The model in `PLOT_MODEL` under `PLOT_SPLIT_METHOD`. With `PLOT_MODEL = None` it is the registry model
with the lowest test MAE under that split method; `PLOT_SPLIT_METHOD` falls back to `static` when
rolling is switched off. Two panels: the local Monday–Sunday week of the test window that contains the
**highest** actual grid-load hour, and the one that contains the **lowest**, clipped to the test
window. Change `PLOT_MODEL` in [§1.3](#sec-1-3) and re-run to see another model.

````python
# Registry rows, best test MAE first; static when PLOT_SPLIT_METHOD has no rows
registry_rows = [row for row in SCOREBOARD_A.index if row[0] in MODELS]
has_split = any(split == PLOT_SPLIT_METHOD for _, split in registry_rows)
plot_split = PLOT_SPLIT_METHOD if has_split else "static"
if PLOT_MODEL is None:
    plot_key = next(model for model, split in registry_rows if split == plot_split)
    picked_by = (
        "  (PLOT_MODEL = None -> lowest test MAE among registry rows "
        f"under {plot_split})"
    )
else:
    plot_key, picked_by = PLOT_MODEL, ""
plot_row = (plot_key, plot_split)
if plot_row not in ROWS:
    raise ValueError(
        f"{plot_row} has no test-year forecast to plot; "
        "pick another PLOT_MODEL or PLOT_SPLIT_METHOD"
    )

band = FORECASTS[plot_row]
label, color = MODELS[plot_key]["label"], MODELS[plot_key]["color"]
test_actual = actual[test_hours]
peaks = {"highest": test_actual.idxmax(), "lowest": test_actual.idxmin()}


def week_of(stamp):
    """Test hours of the local Monday–Sunday week containing `stamp`."""
    monday = stamp.normalize() - pd.Timedelta(days=stamp.dayofweek)
    in_week = (test_hours >= monday) & (test_hours < monday + pd.Timedelta(days=7))
    return test_hours[in_week]


print(f"plotted: {plot_key} / {plot_split}{picked_by}")
fig, axes = plt.subplots(2, 1, figsize=(14, 9))
for ax, (which, peak) in zip(axes, peaks.items()):
    hours = week_of(peak)
    peak_value = f"{test_actual[peak]:,.0f} MWh"
    week_range = f"{hours[0]:%Y-%m-%d} .. {hours[-1]:%Y-%m-%d}"
    peak_hour = f"{peak:%Y-%m-%d %H:%M}"
    print(f"  {which} actual hour {peak_hour} ({peak_value}) -> week {week_range}")
    ax.fill_between(
        hours,
        band.loc[hours, "lower"],
        band.loc[hours, "upper"],
        color=color,
        alpha=0.18,
        linewidth=0,
        label=f"{label}: {INTERVAL['level']:.0%} band",
    )
    ax.plot(
        hours,
        actual[hours],
        color=FIXED["actual"]["color"],
        linewidth=2,
        label=FIXED["actual"]["label"],
    )
    ax.plot(
        hours,
        ROWS[SMARD_ROW][hours],
        color=FIXED["smard"]["color"],
        linewidth=1.6,
        linestyle="--",
        label=FIXED["smard"]["label"],
    )
    ax.plot(
        hours,
        band.loc[hours, "forecast"],
        color=color,
        linewidth=1.8,
        label=f"{label} ({plot_split})",
    )
    ax.axhline(0, color="0.75", linewidth=0.8, zorder=0)
    title = (
        f"Week with the {which} actual grid load "
        f"({peak_value}, {peak:%a %d %b %Y %H:%M})"
    )
    style_timeseries(ax, title, "MWh")
    ax.xaxis.set_major_locator(mdates.DayLocator())
    ax.xaxis.set_minor_locator(mdates.HourLocator(byhour=[6, 12, 18]))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%a %d %b"))
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", ncol=4, frameon=False)
fig.suptitle(
    f"{label} ({plot_split} split) vs. SMARD's day-ahead grid-load forecast, test year",
    fontsize=16,
)
fig.tight_layout(rect=(0, 0.04, 1, 1))
plt.show()
````

| File | Grain | Columns |

````python
EXPORT_DIR = DATA_DIR / "models"
ERROR_COLUMN = f"err_{TARGET}"   # forecast − actual, e.g. err_grid_load
EXPORT_PATHS = {
    "hourly": EXPORT_DIR / "model_load_forecast_errors_hourly.csv",
    "scoreboard": EXPORT_DIR / "model_load_scoreboard.csv",
}
HOURLY_COLUMNS = [
    "timestamp",
    "model",
    "split_method",
    TARGET,
    "forecast",
    "lower",
    "upper",
    ERROR_COLUMN,
]

# One row per test hour, model and split method; SMARD is not in this file
hourly = pd.concat(
    FORECASTS, names=["model", "split_method", "timestamp"]
).reset_index()
hourly[TARGET] = actual[hourly["timestamp"]].to_numpy()
hourly[ERROR_COLUMN] = hourly["forecast"] - hourly[TARGET]
EXPORT_HOURLY = hourly[HOURLY_COLUMNS]
EXPORT_SCOREBOARD = SCOREBOARD[SCOREBOARD_COLUMNS]

row_names = ", ".join(f"{model} / {split}" for model, split in FORECASTS)
print(
    f"model_forecast_errors_hourly (in memory): {len(EXPORT_HOURLY):,} rows = "
    f"{len(test_hours):,} test hours x {len(FORECASTS)} rows ({row_names})"
)
display(EXPORT_HOURLY.head(3))
scored_rows = EXPORT_SCOREBOARD[["model", "split_method"]].drop_duplicates()
print(
    f"model_scoreboard (in memory): {len(EXPORT_SCOREBOARD):,} values "
    f"for {len(scored_rows)} rows"
)
display(EXPORT_SCOREBOARD.head(3))
````

## <a id="sec-9"></a>9 Closing checks and conclusion

### 9.1 Closing self-check

Runs against the in-memory export frames. With `EXPORT_ENABLED` on, it also reads both files back
with a bare `pd.read_csv` and repeats the export checks on the files as written.

````python
def export_checks(hourly, board, source):
    """The export checks on one hourly / scoreboard pair; `source` names it in any failure."""
    assert list(hourly.columns) == HOURLY_COLUMNS and list(board.columns) == SCOREBOARD_COLUMNS, f"{source}: columns"
    error = hourly["forecast"] - hourly[TARGET]
    assert (hourly[ERROR_COLUMN] - error).abs().max() <= 1e-6, f"{source}: {ERROR_COLUMN} != forecast − actual"
    assert (hourly[ERROR_COLUMN].isna() == hourly["forecast"].isna()).all(), f"{source}: error where no forecast"

    # Scoreboard MAE and RMSE against a direct recomputation from the hourly export, on the common hours.
    values = board.set_index(["model", "split_method", "table", "metric"])["value"]
    on_common = pd.to_datetime(hourly["timestamp"]).isin(COMMON).to_numpy()
    for (model, split), rows in hourly[on_common].groupby(["model", "split_method"]):
        if (model, split, "accuracy", "MAE") not in values.index:
            continue   # a row without forecasts has no scoreboard entry
        e = rows[ERROR_COLUMN]
        assert len(e) == len(COMMON) and e.notna().all(), f"{source}: {model}/{split} misses common hours"
        assert abs(values[(model, split, "accuracy", "MAE")] - e.abs().mean()) <= 1e-6, f"{source}: {model}/{split} MAE"
        assert abs(values[(model, split, "accuracy", "RMSE")] - np.sqrt((e ** 2).mean())) <= 1e-6, f"{source}: {model}/{split} RMSE"

    # SMARD's MAE against a direct recomputation from its own errors file.
    smard_file = pd.read_csv(SMARD_ERRORS)
    smard_file["timestamp"] = pd.to_datetime(smard_file["timestamp"], format="%Y-%m-%d %H:%M:%S")
    direct = smard_file.set_index("timestamp").loc[COMMON, BENCHMARK_ERROR].abs().mean()
    assert abs(values[("smard", "none", "accuracy", "MAE")] - direct) <= 1e-6, f"{source}: SMARD MAE"
    assert "smard" not in set(hourly["model"]), f"{source}: SMARD belongs in the metrics file, not the hourly export"


checks = []
assert LEAKAGE_PASSED
checks.append("leakage test by truncation passed (§5.4)")

assert not VAL_DAYS.intersection(TEST_DAYS).size and VAL_DAYS[-1] < TEST_DAYS[0]
assert all(result["tuning"]["forecast"].index.equals(hours_of(VAL_DAYS)) for result in RESULTS.values())
checks.append("test and validation windows do not overlap; no tuning result uses a test hour")

export_checks(EXPORT_HOURLY, EXPORT_SCOREBOARD, "in-memory export")
checks.append(f"export frames: {ERROR_COLUMN} = forecast − actual; scoreboard MAE / RMSE and SMARD's MAE recomputed")

if EXPORT_ENABLED:
    export_checks(pd.read_csv(EXPORT_PATHS["hourly"]), pd.read_csv(EXPORT_PATHS["scoreboard"]), "exported files")
    checks.append("exported files read back with a bare pd.read_csv: the same checks pass")

model_inputs = set(feature_columns()) | set(LINEAR_INPUTS) | {column for columns in FEATURE_GROUPS.values() for column in columns}
for model_key, m in MODELS.items():
    if m["family"] == "sarimax":
        model_inputs |= set(sarimax_exog(time_series.index[:1], m["params"]).columns)
assert "fc_residual_load" not in model_inputs
assert BENCHMARK not in model_inputs, f"{BENCHMARK} is the benchmark and must never be a model input"
checks.append(f"{BENCHMARK} (the benchmark) and fc_residual_load are no model's input")

assert list(time_series.columns) == SERIES + DERIVED, list(time_series.columns)
assert (len(time_series), time_series.index.min(), time_series.index.max()) == (LOADED["rows"], LOADED["start"], LOADED["end"])
assert all(pd.api.types.is_float_dtype(time_series[c]) for c in SERIES)
checks.append(f"time_series still carries exactly SERIES + DERIVED ({LOADED['rows']:,} rows, unchanged since loading)")

print("closing self-check passed")
for check in checks:
    print(f"  - {check}")
print(f"  export {'written' if EXPORT_ENABLED else 'skipped (EXPORT_ENABLED = False)'}; fitting took {FITTING_SECONDS / 60:,.1f} min")
````

## <a id="sec-10"></a>10 Residual load from our load forecast

Our load forecasts become residual-load forecasts the way SMARD builds its own:

```
our residual load   = our grid-load forecast − fc_gen_wind_solar
SMARD residual load = fc_grid_load           − fc_gen_wind_solar   (= fc_residual_load, to SMARD's 0.01 MWh rounding)
```

- **Both use the same wind + solar forecast**, SMARD's `fc_gen_wind_solar`. The comparison therefore
  isolates the load side: per hour, `our residual error − SMARD's residual error = our load error −
  SMARD's load error`. The MAE still differs from §7's load MAE, because a load error can cancel or add to
  the wind + solar error of the same hour.
- **Rows:** every row of §7 (each registry model × split method and seasonal naive), converted this way,
  plus SMARD's `fc_residual_load`.
- **Hours:** §7's common hours, restricted to hours with an actual `residual_load`, a
  `fc_gen_wind_solar` and a `fc_residual_load`. SMARD's error is cross-checked against
  `err_residual_load` in `smard_forecast_errors_hourly.csv`.
- **Tails:** the same bins as §7.3, but on the **actual residual load** of the test window, so
  `low_extreme` / `high_extreme` are the project's two risk cases. Their edges are the P1 / P25 / P75 /
  P99 of the test window, as in `regression-models-claude.ipynb` §7.3, so the two notebooks' tail
  tables can be read side by side.
- **No band:** a residual band would also need the wind + solar uncertainty, which the load band does
  not contain.

### 10.1 Accuracy


````python
RL_ACTUAL = time_series["residual_load"]
RL_RENEWABLES = time_series["fc_gen_wind_solar"]  # SMARD's wind + solar forecast, as in SMARD's own residual forecast
SMARD_RL = time_series["fc_residual_load"]
IDENTITY_TOLERANCE = 0.05  # MWh: SMARD rounds its forecasts to 0.01 MWh

# SMARD's residual forecast is its load forecast minus the same wind + solar forecast
identity_gap = float((SMARD_RL - (time_series[BENCHMARK] - RL_RENEWABLES)).abs().max())
assert identity_gap <= IDENTITY_TOLERANCE, f"fc_residual_load != {BENCHMARK} − fc_gen_wind_solar by up to {identity_gap} MWh"

# Residual-load forecast per row: the load forecast minus fc_gen_wind_solar; SMARD's own forecast
RL_ROWS = {
    row: forecast - RL_RENEWABLES[forecast.index]
    for row, forecast in ROWS.items()
    if row != SMARD_ROW
}
RL_ROWS[SMARD_ROW] = SMARD_RL[test_hours]

# Common hours: §7's, with an actual residual load and both SMARD inputs
rl_observed = (RL_ACTUAL[COMMON].notna() & RL_RENEWABLES[COMMON].notna() & SMARD_RL[COMMON].notna()).to_numpy()
RL_COMMON = COMMON[rl_observed]
RL_ERRORS = {row: forecast[RL_COMMON] - RL_ACTUAL[RL_COMMON] for row, forecast in RL_ROWS.items()}

# SMARD re-scored against its hourly errors file
rl_cross_check = float((RL_ERRORS[SMARD_ROW] - smard_errors.loc[RL_COMMON, "err_residual_load"]).abs().max())
assert rl_cross_check <= 1e-6, f"SMARD's residual error disagrees with the errors file by {rl_cross_check} MWh"

# Accuracy per row, with the row's grid-load MAE from Scoreboard A next to it
rl_month_of = RL_COMMON.to_period("M")
rl_monthly_mae = {row: error.abs().groupby(rl_month_of).mean() for row, error in RL_ERRORS.items()}
RL_MAE_SMARD = RL_ERRORS[SMARD_ROW].abs().mean()
rl_accuracy = {}
for row, error in RL_ERRORS.items():
    mae = error.abs().mean()
    metrics = {
        "MAE": mae,
        "RMSE": np.sqrt((error**2).mean()),
        "bias": error.mean(),
        "grid-load MAE (§7.2)": SCOREBOARD_A.loc[row, "MAE"],
    }
    if row != SMARD_ROW:
        beats = rl_monthly_mae[row][FULL_MONTHS] < rl_monthly_mae[SMARD_ROW][FULL_MONTHS]
        metrics["skill vs SMARD (%)"] = 100 * (1 - mae / RL_MAE_SMARD)
        metrics["months beating SMARD"] = int(beats.sum())
    rl_accuracy[row] = metrics
RL_SCOREBOARD = pd.DataFrame(rl_accuracy).T.rename_axis(["model", "split_method"]).sort_values("MAE")

print(
    f"Residual load from our load forecast vs. SMARD's fc_residual_load: {len(RL_COMMON):,} common test hours "
    f"({TEST_DAYS[0]:%Y-%m-%d} .. {TEST_DAYS[-1]:%Y-%m-%d}); MWh; {len(FULL_MONTHS)} full calendar months"
)
print(f"  fc_residual_load = {BENCHMARK} − fc_gen_wind_solar within {identity_gap:.2f} MWh; SMARD cross-check {rl_cross_check:.1e} MWh")
rl_formats = {
    "MAE": "{:,.0f}",
    "RMSE": "{:,.0f}",
    "bias": "{:+,.0f}",
    "grid-load MAE (§7.2)": "{:,.0f}",
    "skill vs SMARD (%)": "{:+.1f}",
    "months beating SMARD": "{:.0f}",
}
display(RL_SCOREBOARD.style.format(rl_formats, na_rep="-"))
````

### 10.2 Extremes

MAE, bias and skill vs SMARD in the tails of the **actual residual load**, and the value error of each
day's residual-load maximum and minimum. A day counts if its common hours cover at least
`DAY_COMPLETENESS` of its expected hours, as in §7.3. Only the view binned by the actual is shown here:
it is the one where every row is scored on the same hours.


````python
RL_EDGES = RL_ACTUAL[test_hours].quantile(TAIL_LEVELS)
RL_P1, RL_P25, RL_P75, RL_P99 = RL_EDGES.to_numpy()
rl_actual = RL_ACTUAL[RL_COMMON]
rl_bins = pd.Series(
    np.select(
        [rl_actual <= RL_P1, (rl_actual >= RL_P25) & (rl_actual <= RL_P75), rl_actual > RL_P99],
        BINS,
        default="other",
    ),
    index=RL_COMMON,
)

# Days whose common hours cover at least DAY_COMPLETENESS of their expected hours
rl_day = RL_COMMON.normalize()
rl_per_day = pd.Series(1, index=RL_COMMON).groupby(rl_day).sum()
rl_expected = FORECAST_SETTING["target_hours"].map(len).reindex(rl_per_day.index)
RL_DAYS_SCORED = rl_per_day.index[(rl_per_day >= DAY_COMPLETENESS * rl_expected).to_numpy()]
rl_scored = rl_day.isin(RL_DAYS_SCORED)


def rl_day_value_error(row, agg):
    """forecast − actual of each scored day's residual-load maximum or minimum."""
    day = rl_day[rl_scored]
    forecast = RL_ROWS[row][RL_COMMON][rl_scored].groupby(day).agg(agg)
    observed = rl_actual[rl_scored].groupby(day).agg(agg)
    return forecast - observed


rl_extremes = {}
for row, error in RL_ERRORS.items():
    reference = None if row == SMARD_ROW else RL_ERRORS[SMARD_ROW]
    for bin_name in BINS:
        in_bin = (rl_bins == bin_name).to_numpy()
        smard_part = None if reference is None else reference[in_bin]
        for metric, value in group_metrics(error[in_bin], smard_part).items():
            rl_extremes[(*row, bin_name, metric)] = value
    for agg in ["max", "min"]:
        day_error = rl_day_value_error(row, agg)
        smard_day = None if row == SMARD_ROW else rl_day_value_error(SMARD_ROW, agg)
        for metric, value in group_metrics(day_error, smard_day).items():
            rl_extremes[(*row, f"day_{agg}", metric)] = value

RL_EXTREMES = pd.Series(rl_extremes).unstack([2, 3]).loc[RL_SCOREBOARD.index]
RL_EXTREMES = RL_EXTREMES.rename(columns={"skill_pct": "skill vs SMARD (%)"}, level=1)
RL_EXTREMES = RL_EXTREMES.reindex(columns=[*BINS, "day_max", "day_min"], level=0)

rl_bin_hours = rl_bins.value_counts()
edges_text = ", ".join(f"P{level * 100:g} = {edge:,.0f}" for level, edge in RL_EDGES.items())
print("Tail-bin edges: quantiles of the actual residual load over the test window (MWh)")
print(f"  {edges_text}")
print(
    "hours: " + ", ".join(f"{name} {rl_bin_hours.get(name, 0):,}" for name in BINS)
    + f"; day maximum / minimum over {len(RL_DAYS_SCORED)} days"
)
extreme_column_formats = {
    column: extreme_formats[column[1]] for column in RL_EXTREMES.columns
}
display(RL_EXTREMES.style.format(extreme_column_formats, na_rep="—"))
````
